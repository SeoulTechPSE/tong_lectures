# String Theory Ch 7 — Figure Notebook
**Low-Energy Effective Action**

```python
FIG_DIR = 'figures/ch07'

import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import os

BG='#1e1e2e';FG='#cdd6f4';EDGE='#45475a'
SKYB='#89c4f4';YELL='#f9c74f';GREEN='#90ee90'
RED='#e07070';PURP='#c792ea';ORAN='#e8a838';TEAL='#2ec4b6'

plt.rcParams.update({
    'figure.facecolor':BG,'axes.facecolor':BG,
    'text.color':FG,'axes.labelcolor':FG,
    'xtick.color':FG,'ytick.color':FG,
    'axes.edgecolor':EDGE,'grid.color':EDGE,'grid.alpha':0.25,
    'font.size':12,'axes.titlesize':13,'axes.titleweight':'bold',
    'axes.titlecolor':SKYB,'legend.framealpha':0.3,
    'legend.edgecolor':EDGE,'lines.linewidth':2.2,
    'axes.linewidth':1.2,'figure.dpi':110,
})
os.makedirs(FIG_DIR, exist_ok=True)

def save(name, fig):
    fig.savefig(f'{FIG_DIR}/{name}', dpi=150, bbox_inches='tight', facecolor=BG)
    plt.show(); print(f'Saved: {FIG_DIR}/{name}')

print('Setup complete.')
```

```python

# ── fig01: Massless fields and effective action ───────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(12, 5.5))
fig.suptitle('Ch 7 · Low-Energy Effective Action: Massless Fields',
             color=SKYB, fontsize=13)

ax = axes[0]; ax.set_facecolor(BG); ax.axis('off')
ax.set_title(r'$S_{\rm eff}$: beta function origin', color=SKYB)
items = [
    (0.5, 0.92, r'Nonlinear sigma model on curved background:', SKYB),
    (0.5, 0.81, r'$S_\sigma=\frac{1}{4\pi\alpha\prime}\int d^2\sigma\sqrt{h}[h^{ab}G_{\mu\nu}\partial_a X^\mu\partial_b X^\nu$', FG),
    (0.5, 0.70, r'$+\epsilon^{ab}B_{\mu\nu}\partial_a X^\mu\partial_b X^\nu+\alpha\prime R^{(2)}\Phi]$', FG),
    (0.5, 0.57, r'Weyl invariance: $\beta^G=\beta^B=\beta^\Phi=0$', YELL),
    (0.5, 0.45, r'$\beta^G_{\mu\nu}=\alpha\prime R_{\mu\nu}-\frac{\alpha\prime}{4}H_{\mu\rho\sigma}H_\nu{}^{\rho\sigma}+2\alpha\prime\nabla_\mu\nabla_\nu\Phi+\cdots$', GREEN),
    (0.5, 0.33, r'$\Rightarrow$ equations of motion of $S_{\rm eff}$', GREEN),
    (0.5, 0.21, r'$S_{\rm eff}=\frac{1}{2\kappa^2}\int d^Dx\sqrt{G}\,e^{-2\Phi}[R+4(\nabla\Phi)^2-\frac{H^2}{12}]$', RED),
    (0.5, 0.08, r'String frame: graviton, dilaton, Kalb-Ramond', TEAL),
]
for x_,y_,txt,col in items:
    ax.text(x_,y_,txt,ha='center',va='center',fontsize=10.5,color=col,
            transform=ax.transAxes,
            bbox=dict(boxstyle='round,pad=0.18',fc=BG,ec=col,alpha=0.4))

# Right: massless field spectrum diagram
ax2 = axes[1]; ax2.set_facecolor(BG); ax2.axis('off')
ax2.set_title('Massless closed string fields', color=SKYB)

fields = [
    (0.25, 0.75, r'$G_{\mu\nu}$', 'Graviton\n(spin 2)', GREEN),
    (0.50, 0.75, r'$B_{\mu\nu}$', 'Kalb-Ramond\n(2-form)', PURP),
    (0.75, 0.75, r'$\Phi$',       'Dilaton\n(scalar)', YELL),
]
ax2.text(0.5, 0.95, r'From $\alpha^i_{-1}\tilde\alpha^j_{-1}|0;p\rangle$ at $m^2=0$:',
         ha='center', fontsize=11, color=SKYB, transform=ax2.transAxes)
for x_, y_, sym, desc, col in fields:
    circ = plt.Circle((x_, y_), 0.12, color=col, alpha=0.25,
                       transform=ax2.transAxes, clip_on=False)
    ax2.add_patch(circ)
    ax2.text(x_, y_+0.01, sym, ha='center', va='center', fontsize=16,
             color=col, transform=ax2.transAxes, fontweight='bold')
    ax2.text(x_, y_-0.2, desc, ha='center', va='center', fontsize=10,
             color=col, transform=ax2.transAxes)

# Arrows showing origin
for x_ in [0.25, 0.50, 0.75]:
    ax2.annotate('', xy=(x_, 0.57), xytext=(0.5, 0.92),
                 xycoords='axes fraction', textcoords='axes fraction',
                 arrowprops=dict(arrowstyle='->', color=EDGE, lw=1.2))

# Physical content
ax2.text(0.5, 0.38,
         r'String theory $\Rightarrow$ Einstein gravity (inevitably!)',
         ha='center', fontsize=11, color=GREEN, transform=ax2.transAxes,
         bbox=dict(boxstyle='round', fc=BG, ec=GREEN, alpha=0.8))
ax2.text(0.5, 0.25,
         r'String coupling: $g_s=e^{\langle\Phi\rangle}$ (dynamical)',
         ha='center', fontsize=11, color=YELL, transform=ax2.transAxes,
         bbox=dict(boxstyle='round', fc=BG, ec=YELL, alpha=0.8))
ax2.text(0.5, 0.12,
         r'NS5-brane: magnetic dual of fundamental string',
         ha='center', fontsize=11, color=PURP, transform=ax2.transAxes,
         bbox=dict(boxstyle='round', fc=BG, ec=PURP, alpha=0.8))

plt.tight_layout()
save('fig01_effective_action_fields.png', fig)
```

```python

# ── fig02: alpha' expansion and dilaton potential ─────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(12, 5.5))
fig.suptitle(r"Ch 7 · $\alpha'$ Expansion and Dilaton Dynamics",
             color=SKYB, fontsize=13)

# Left: alpha' expansion schematic
ax = axes[0]; ax.set_facecolor(BG); ax.axis('off')
ax.set_title(r"$\alpha'$ expansion of $S_{\rm eff}$", color=SKYB)
items = [
    (0.5, 0.92, r'$S=S_{\rm EH}+\alpha^{\prime}(\cdots)+{\alpha^{\prime}}^{3}(\cdots)+\cdots$', SKYB),
    (0.5, 0.80, r'Order ${\alpha^{\prime}}^{0}$: Einstein-Hilbert', GREEN),
    (0.5, 0.69, r'$\frac{1}{2\kappa^2}\int\sqrt{G}[R+4(\nabla\Phi)^2-\frac{H^2}{12}]$', GREEN),
    (0.5, 0.57, r'Order ${\alpha^{\prime}}^{1}$: Gauss-Bonnet', YELL),
    (0.5, 0.46, r'$+\alpha^{\prime}[R_{\mu\nu\rho\sigma}^2-4R_{\mu\nu}^2+R^2]+\cdots$', YELL),
    (0.5, 0.34, r'Order ${\alpha^{\prime}}^{3}$: $R^4$ terms', ORAN),
    (0.5, 0.23, r'First correction to $\mathcal{N}=8$ supergravity', ORAN),
    (0.5, 0.11, r'Valid for $\mathcal{R}\alpha^{\prime}\ll 1$ (low curvature)', TEAL),
]
for x_,y_,txt,col in items:
    ax.text(x_,y_,txt,ha='center',va='center',fontsize=11,color=col,
            transform=ax.transAxes,
            bbox=dict(boxstyle='round,pad=0.18',fc=BG,ec=col,alpha=0.4))

# Right: dilaton potential and string coupling
ax2 = axes[1]; ax2.set_facecolor(BG)
ax2.set_title(r'Dilaton $\Phi$ and string coupling $g_s=e^\Phi$', color=SKYB)
Phi = np.linspace(-3, 3, 300)
gs  = np.exp(Phi)
# Perturbative regime: g_s << 1  <-> Phi << 0
# Non-perturbative: instanton V ~ exp(-1/g_s^2)
V_pert = np.zeros_like(Phi)  # flat at tree level
V_np   = np.exp(-np.exp(-2*Phi))  # schematic non-perturbative

ax2_twin = ax2.twinx()
ax2_twin.set_facecolor(BG)
ax2.plot(Phi, gs, color=YELL, lw=2.5, label=r'$g_s=e^\Phi$')
ax2.axvline(0, color=EDGE, lw=1, ls='--')
ax2.axhline(1, color=EDGE, lw=1, ls=':', alpha=0.6)
ax2.fill_between(Phi, 0, gs, where=(Phi<0), color=SKYB, alpha=0.07,
                 label='Perturbative: $g_s<1$')
ax2.fill_between(Phi, 0, gs, where=(Phi>0), color=RED, alpha=0.07,
                 label='Strong coupling: $g_s>1$')
ax2_twin.plot(Phi, V_np, color=GREEN, lw=2.5, ls='--',
              label=r'$V(\Phi)\sim e^{-1/g_s^2}$ (non-pert.)')
ax2_twin.set_ylabel(r'Non-perturbative potential (arb.)', color=GREEN)
ax2_twin.tick_params(axis='y', colors=GREEN)

ax2.set_xlabel(r'Dilaton $\Phi$')
ax2.set_ylabel(r'String coupling $g_s$')
lines1, labels1 = ax2.get_legend_handles_labels()
lines2, labels2 = ax2_twin.get_legend_handles_labels()
ax2.legend(lines1+lines2, labels1+labels2, fontsize=9, loc='upper left')
ax2.text(0.02, 0.2, r'Moduli stabilisation needed!',
         transform=ax2.transAxes, fontsize=10, color=RED,
         bbox=dict(boxstyle='round', fc=BG, ec=RED, alpha=0.8))
ax2.grid(True, alpha=0.2)

plt.tight_layout()
save('fig02_alpha_prime_dilaton.png', fig)
```
