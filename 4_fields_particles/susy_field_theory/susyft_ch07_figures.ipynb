# SUSYFT CH07 Figures
**More Gauge Dynamics and Dynamical SUSY Breaking**

```python

import numpy as np, matplotlib.pyplot as plt, os
BG='#1e1e2e';FG='#cdd6f4';EDGE='#45475a'
GOLD='#e8b84b';BLUE='#7ec8e3';PURP='#c792ea'
GREEN='#90ee90';RED='#e07070';ORAN='#e8a838';TEAL='#2ec4b6'
plt.rcParams.update({'figure.facecolor':BG,'axes.facecolor':BG,'text.color':FG,
    'axes.labelcolor':FG,'xtick.color':FG,'ytick.color':FG,'axes.edgecolor':EDGE,
    'grid.color':EDGE,'grid.alpha':0.3,'font.size':12,'axes.titlesize':13,
    'axes.titleweight':'bold','axes.titlecolor':GOLD,'lines.linewidth':2.2,
    'axes.linewidth':1.2,'figure.dpi':110})
OUTDIR='figures/ch07'; os.makedirs(OUTDIR,exist_ok=True)
def save(name,fig=None):
    f=fig or plt.gcf()
    p=f'{OUTDIR}/{name}'; f.savefig(p,dpi=180,bbox_inches='tight',facecolor=BG)
    print(f'saved->{p}'); plt.show()
print('Setup ch07')
```

```python
fig, ax = plt.subplots(figsize=(11, 5.5))
fig.suptitle(r'Other Gauge Groups, Chiral Theories, $a$-Maximisation',
             color=GOLD, fontsize=12)
ax.set_facecolor(BG); ax.axis('off')

rows = [
    (r'$Sp(N)$ gauge theory: similar Seiberg duality structure to $SU(N)$',              GOLD),
    (r'$Sp(N)$ dual: $Sp(N_f-N-2)$ + $N_f$ quarks + mesons $M$, $W_{mag}=Mqq$',        GOLD),
    (r'$SU(2)=Sp(1)$: special case, $N_f=1$ quantum moduli space',                       BLUE),
    (r'Intriligator--Pouliot duality: $Sp(N)$ with antisymmetric tensor',                BLUE),
    (r'Chiral gauge theory: matter in complex reps (no vector-like version)',             TEAL),
    (r'$SU(N)$ with symmetric $S$ + antifundamentals: chiral content',                   TEAL),
    (r'Pouliot--Strassler: $SU(N)$ chiral duality (1995)',                               PURP),
    (r'$a$-maximisation: find exact $U(1)_R$ at IR fixed point',                         ORAN),
    (r'Maximise $a=\frac{3}{32}(3\,\mathrm{Tr}R^3 - \mathrm{Tr}R)$ over trial $R$-charges', ORAN),
    (r'Konishi anomaly: $\bar{D}^2(\Phi_i^\dagger e^{2V}\Phi_i) = \Phi_i\partial_i W + \mathrm{quant.}$', RED),
]
ys = np.linspace(0.92, 0.06, len(rows))

for (txt, col), y in zip(rows, ys):
    ax.text(0.5, y, txt, ha='center', va='center', fontsize=11, color=col,
            transform=ax.transAxes,
            bbox=dict(boxstyle='round,pad=0.2', fc=BG, ec=col, alpha=0.5))

plt.tight_layout()
save('fig01_other_groups.png', fig)
```

```python
fig, ax = plt.subplots(figsize=(11, 5.5))
fig.suptitle('Dynamical Supersymmetry Breaking', color=GOLD, fontsize=12)
ax.set_facecolor(BG); ax.axis('off')

rows = [
    (r'Dynamical SUSY breaking: $E_0 > 0$ generated non-perturbatively',                          GOLD),
    (r'Scale: $m_{SUSY} \sim \Lambda\,e^{-8\pi^2/b_0 g^2}$ — exponentially small',               RED),
    (r'Natural hierarchy: $m_{SUSY} \ll M_{Pl}$ without fine-tuning',                             GREEN),
    (r'Requirements: $\Delta_W = 0$ AND appropriate $U(1)_R$ symmetry',                           PURP),
    (r'$SU(3)\times SU(2)$ model (Affleck--Dine--Seiberg, Izawa--Yanagida):',                     BLUE),
    (r'$W_{ADS}$ from $SU(3)$ instantons + tree-level $SU(2)$ coupling',                          BLUE),
    (r'Result: isolated SUSY-breaking minimum, $\Delta_W = 0$',                                   ORAN),
    (r'Quantum moduli space: classical flat direction lifted by dynamics',                          TEAL),
    (r'Goldstino $\tilde{G}$: massless fermion from SUSY breaking, eaten by gravitino',            ORAN),
    (r'Open question: realistic mediation of $m_{SUSY}$ to the Standard Model',                   FG),
]
ys = np.linspace(0.92, 0.06, len(rows))

for (txt, col), y in zip(rows, ys):
    ax.text(0.5, y, txt, ha='center', va='center', fontsize=11, color=col,
            transform=ax.transAxes,
            bbox=dict(boxstyle='round,pad=0.2', fc=BG, ec=col, alpha=0.5))

plt.tight_layout()
save('fig02_dynamical_susy.png', fig)
```
