# SUSYFT CH06 Figures
**Supersymmetric QCD — Phases and Seiberg Duality**

```python

import numpy as np, matplotlib.pyplot as plt, os
BG='#1e1e2e';FG='#cdd6f4';EDGE='#45475a'
GOLD='#e8b84b';BLUE='#7ec8e3';PURP='#c792ea'
GREEN='#90ee90';RED='#e07070';ORAN='#e8a838';TEAL='#2ec4b6'
plt.rcParams.update({'figure.facecolor':BG,'axes.facecolor':BG,'text.color':FG,
    'axes.labelcolor':FG,'xtick.color':FG,'ytick.color':FG,'axes.edgecolor':EDGE,
    'grid.color':EDGE,'grid.alpha':0.3,'font.size':12,'axes.titlesize':13,
    'axes.titleweight':'bold','axes.titlecolor':GOLD,'lines.linewidth':2.2,
    'axes.linewidth':1.2,'figure.dpi':110})
OUTDIR='figures/ch06'; os.makedirs(OUTDIR,exist_ok=True)
def save(name,fig=None):
    f=fig or plt.gcf()
    p=f'{OUTDIR}/{name}'; f.savefig(p,dpi=180,bbox_inches='tight',facecolor=BG)
    print(f'saved->{p}'); plt.show()
print('Setup ch06')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5.5))
fig.suptitle('SQCD Phase Diagram and Seiberg Duality', color=GOLD, fontsize=12)

# ── 왼쪽: 위상 다이어그램 ──────────────────────────────────────────────────
ax = axes[0]; ax.set_facecolor(BG); ax.axis('off')
ax.set_title(r'Phases vs $N_f/N_c$', color=GOLD)
ax.set_xlim(0, 1.5); ax.set_ylim(-0.18, 1.05)

regions = [
    #  x0    x1    label                          color  xm      y_label
    (0.00, 0.33, 'Confining\nADS runaway',        RED,   0.165,  0.50),
    (0.33, 0.50, '$N_f=N_c, N_c+1$\nspecial',     ORAN,  0.415,  0.75),  # 좁으므로 위로
    (0.50, 0.67, 'Free magnetic\n(Seiberg dual)',  PURP,  0.585,  0.30),  # 좁으므로 아래로
    (0.67, 1.00, 'Conformal\nwindow',              TEAL,  0.835,  0.60),
    (1.00, 1.35, 'IR free\n(electric)',             GREEN, 1.175,  0.50),
]

for x0, x1, lbl, col, xm, ly in regions:
    ax.fill_between([x0, x1], [0, 0], [1, 1], alpha=0.18, color=col)
    ax.axvline(x0, color=col, lw=1, ls='--', alpha=0.5)
    ax.text(xm, ly, lbl, ha='center', va='center', fontsize=9,
            color=col, fontweight='bold',
            bbox=dict(boxstyle='round,pad=0.2', fc=BG, ec=col, alpha=0.85))

for xv, lbl in [(0.33, '1'), (0.67, r'$\frac{3}{2}$'), (1.00, '3')]:
    ax.text(xv, -0.10, lbl, ha='center', fontsize=10, color=FG)
ax.text(0.7, -0.15, r'$N_f / N_c$', ha='center', fontsize=10, color=FG)

# ── 오른쪽: Seiberg 이중성 ────────────────────────────────────────────────
ax2 = axes[1]; ax2.set_facecolor(BG); ax2.axis('off')
ax2.set_title('Seiberg Duality (Seiberg 1994)', color=GOLD)

rows = [
    (r'Electric: $SU(N_c)$ + $N_f$ quarks $Q$, $\tilde{Q}$',                          GOLD),
    (r'Magnetic: $SU(N_f-N_c)$ + $N_f$ quarks $q,\tilde{q}$ + mesons $M=Q\tilde{Q}$', BLUE),
    (r'Magnetic superpotential: $W_{mag} = M q\tilde{q}$',                            BLUE),
    (r'Strong coupling $\longleftrightarrow$ Weak coupling',                          GOLD),
    (r'Same IR fixed point: same operators, same anomalies',                          GREEN),
    (r"'t Hooft anomaly matching: verified",                                          GREEN),
    (r'Deformations: add mass $\delta W=m\tilde{Q}Q$ — matches on both sides',        TEAL),
    (r'Moduli space: exact agreement',                                                TEAL),
    (r'Range: $N_c < N_f < \frac{3}{2}N_c$ (free magnetic phase)',                    PURP),
    (r'Generalises EM duality: $g \leftrightarrow 1/g$ in SUSY context',              RED),
]
ys = np.linspace(0.93, 0.06, len(rows))
for (txt, col), y in zip(rows, ys):
    ax2.text(0.5, y, txt, ha='center', va='center', fontsize=10, color=col,
             transform=ax2.transAxes,
             bbox=dict(boxstyle='round,pad=0.15', fc=BG, ec=col, alpha=0.5))

plt.tight_layout()
save('fig01_sqcd_seiberg.png', fig)
```

```python
fig, ax = plt.subplots(figsize=(11, 5.5))
fig.suptitle(r'$\mathcal{N}=1$ SYM: Witten Index, Confinement, Gluino Condensate',
             color=GOLD, fontsize=12)
ax.set_facecolor(BG); ax.axis('off')

rows = [
    (r'$\mathcal{N}=1$ SYM: $SU(N_c)$ pure gauge + gaugino $\lambda$', GOLD),
    (r'Witten index: $\Delta_W = N_c$ (Witten 1982)', GREEN),
    (r'$N_c$ supersymmetric vacua, related by $\mathbb{Z}_{N_c}$ center symmetry', GREEN),
    (r'Gluino condensate: $\langle\lambda\lambda\rangle = \Lambda^3 e^{2\pi ik/N_c}$, $k=0,\ldots,N_c-1$', BLUE),
    (r'Discrete chiral symmetry: $\mathbb{Z}_{2N_c}\to\mathbb{Z}_2$ spontaneously broken', PURP),
    (r'Theory confines: mass gap, no massless particles', TEAL),
    (r'Adding quarks ($N_f<N_c$): $\Delta_W = N_c - N_f$', ORAN),
    (r'$N_f = N_c$: $\Delta_W=0$, deformed moduli space, no runaway', RED),
    (r'$N_f = N_c+1$: confinement without $\chi$SB, $\Delta_W=0$ by anomaly matching', ORAN),
    (r'Key: all results protected by SUSY non-renormalisation', GREEN),
]
ys = np.linspace(0.92, 0.06, len(rows))

for (txt, col), y in zip(rows, ys):          # ← (txt, col) 로 언팩
    ax.text(0.5, y, txt, ha='center', va='center', fontsize=11, color=col,
            transform=ax.transAxes,
            bbox=dict(boxstyle='round,pad=0.2', fc=BG, ec=col, alpha=0.5))

plt.tight_layout()
save('fig02_sym_index.png', fig)
```

```python
fig, ax = plt.subplots(figsize=(11, 5.5))
fig.suptitle(r'Conformal Window, $a$-Maximisation, Superconformal Field Theories',
             color=GOLD, fontsize=12)
ax.set_facecolor(BG); ax.axis('off')

rows = [
    (r'Conformal window: $\frac{3}{2}N_c < N_f < 3N_c$', GOLD),
    (r'IR fixed point: $\beta(g^*)=0$ at $g^*\neq 0$ (non-trivial)', GREEN),
    (r'Banks-Zaks fixed point: weakly coupled near $N_f=3N_c$', BLUE),
    (r'Superconformal algebra: adds generators $K_\mu$, $S_\alpha$, dilatation $D$', TEAL),
    (r'$\{Q_\alpha, S^\alpha\} = R + M$ — R-charge determines scaling dimensions', TEAL),
    (r'$a$-maximisation (Intriligator-Wecht 2003): find exact $U(1)_R$', ORAN),
    (r'Maximise $a=\frac{3}{32}(3\,\mathrm{Tr}R^3 - \mathrm{Tr}R)$ over trial R-charges', ORAN),
    (r'Chiral primaries: $\Delta = \frac{3}{2}|R|$ exactly', PURP),
    (r'Unitarity bound: $\Delta\geq 1$ (scalar), $\Delta\geq\frac{3}{2}$ (spinor)', RED),
    (r'Seiberg dual in conformal window: two different SCFT descriptions', GREEN),
]
ys = np.linspace(0.92, 0.06, len(rows))

for (txt, col), y in zip(rows, ys):
    ax.text(0.5, y, txt, ha='center', va='center', fontsize=11, color=col,
            transform=ax.transAxes,
            bbox=dict(boxstyle='round,pad=0.2', fc=BG, ec=col, alpha=0.5))

plt.tight_layout()
save('fig03_conformal_window.png', fig)
```
