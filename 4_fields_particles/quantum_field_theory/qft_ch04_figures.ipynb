## fig01 · Lorentz Representations & Dirac Spectrum (§4.1–4.2)

**Left**: Representations of $\mathrm{SO}(1,3)\cong SU(2)\times SU(2)$. **Centre**: Dirac algebra $\{\gamma^\mu,\gamma^\nu\}=2\eta^{\mu\nu}$, $\gamma^5$. **Right**: Particle ($E>0$) and antiparticle ($E<0$) branches — mass gap $2m$.

```python
import numpy as np
import matplotlib.pyplot as plt
import os

BG='#1e1e2e'; FG='#cdd6f4'; EDGE='#45475a'
BLUE='#7ec8e3'; YELLOW='#f9c74f'; GREEN='#90ee90'
RED='#e07070'; PURPLE='#c792ea'; ORANGE='#ffb347'
plt.rcParams.update({
    'figure.facecolor':BG,'axes.facecolor':BG,'text.color':FG,
    'axes.labelcolor':FG,'xtick.color':FG,'ytick.color':FG,
    'axes.edgecolor':EDGE,'grid.color':EDGE,'grid.alpha':0.3,
    'font.size':12,'axes.titlesize':13,'axes.titleweight':'bold',
    'axes.titlecolor':BLUE,'legend.framealpha':0.3,'legend.edgecolor':EDGE,
})
OUTDIR = 'figures/ch04'; os.makedirs(OUTDIR, exist_ok=True)
def save(name, fig=None):
    f = fig or plt.gcf(); path = f'{OUTDIR}/{name}'
    f.savefig(path, dpi=180, bbox_inches='tight', facecolor=BG)
    print(f'  saved -> {path}'); plt.show()

# ── fig01: Lorentz group representations + Dirac spinor ───────────────────
fig, axes = plt.subplots(1, 3, figsize=(14, 5))
fig.patch.set_facecolor(BG)

# Left: representation diagram (j_L, j_R)
ax = axes[0]
ax.axis('off')
reps = [
    (0.25, 0.82, r'$(0,0)$',              'Scalar',         GREEN,  12),
    (0.25, 0.67, r'$(1/2,0)$',            'Left Weyl',      BLUE,   12),
    (0.25, 0.52, r'$(0,1/2)$',            'Right Weyl',     YELLOW, 12),
    (0.25, 0.37, r'$(1/2,0)\oplus(0,1/2)$','Dirac spinor',  RED,    11),
    (0.25, 0.22, r'$(1/2,1/2)$',          '4-vector',       ORANGE, 12),
    (0.25, 0.08, r'$(1,0)\oplus(0,1)$',   'Field strength', PURPLE, 11),
]
ax.text(0.05, 0.93, r'$(j_L,j_R)$', color=BLUE, fontsize=11,
        fontweight='bold', transform=ax.transAxes)
ax.text(0.60, 0.93, 'Field type', color=BLUE, fontsize=11,
        fontweight='bold', transform=ax.transAxes)
for (xp, yp, rep, name, col, fs) in reps:
    ax.text(xp, yp, rep,  ha='center', color=col,  fontsize=fs, transform=ax.transAxes)
    ax.text(0.72, yp, name, ha='center', color=FG, fontsize=10, transform=ax.transAxes)
for yp in [0.87, 0.74, 0.59, 0.44, 0.29, 0.14]:
    ax.plot([0.02, 0.98], [yp, yp], color=EDGE, lw=0.6, transform=ax.transAxes)
ax.set_title('Lorentz Group Representations\n' + r'$\mathrm{SO}(1,3)\cong SU(2)\times SU(2)$',
             color=BLUE)

# Centre: Dirac gamma matrices (schematic 4x4 structure)
ax2 = axes[1]
ax2.axis('off')
items2 = [
    (0.50, 0.90, 'Dirac Algebra', BLUE, 13, 'bold'),
    (0.50, 0.78, r'$\{\gamma^\mu,\gamma^\nu\}=2\eta^{\mu\nu}\mathbf{1}_{4\times4}$',
     YELLOW, 12, 'normal'),
    (0.50, 0.64, r'Dirac basis:', GREEN, 10, 'bold'),
    (0.50, 0.55, r'$\gamma^0=\begin{pmatrix}\mathbf{1}&0\\0&-\mathbf{1}\end{pmatrix}$',
     FG, 11, 'normal'),
    (0.50, 0.40, r'$\gamma^i=\begin{pmatrix}0&\sigma^i\\-\sigma^i&0\end{pmatrix}$',
     FG, 11, 'normal'),
    (0.50, 0.26, r'$\gamma^5=i\gamma^0\gamma^1\gamma^2\gamma^3$', ORANGE, 11, 'normal'),
    (0.50, 0.14, r'$(\gamma^5)^2=1,\quad\{\gamma^5,\gamma^\mu\}=0$', PURPLE, 10, 'normal'),
]
for (xp, yp, txt, col, fs, fw) in items2:
    ax2.text(xp, yp, txt, ha='center', va='center', color=col,
             fontsize=fs, fontweight=fw, transform=ax2.transAxes)
rect2 = plt.Rectangle((0.03, 0.03), 0.94, 0.94, fill=True,
                       facecolor=GREEN, alpha=0.05,
                       edgecolor=GREEN, lw=1.5, transform=ax2.transAxes)
ax2.add_patch(rect2)
ax2.set_title('Dirac Algebra\n' + r'$\{\gamma^\mu,\gamma^\nu\}=2\eta^{\mu\nu}$', color=BLUE)

# Right: Dirac equation plane-wave solutions (E vs p dispersion)
ax3 = axes[2]
p = np.linspace(-4, 4, 300)
m = 1.0
E_pos =  np.sqrt(p**2 + m**2)
E_neg = -np.sqrt(p**2 + m**2)
ax3.plot(p, E_pos, color=BLUE,   lw=2.5, label=r'$E=+\sqrt{p^2+m^2}$ (particle)')
ax3.plot(p, E_neg, color=RED,    lw=2.5, label=r'$E=-\sqrt{p^2+m^2}$ (antiparticle)')
ax3.fill_between(p, E_neg, -m, color=RED,  alpha=0.10)
ax3.fill_between(p,  m, E_pos,  color=BLUE, alpha=0.10)
ax3.fill_between(p, -m, m, color=EDGE, alpha=0.25)
ax3.text(0, 0, 'Mass gap\n$2m$', ha='center', va='center', color=FG, fontsize=10)
ax3.axhline( m, color=YELLOW, lw=1.2, ls='--', alpha=0.7)
ax3.axhline(-m, color=YELLOW, lw=1.2, ls='--', alpha=0.7)
ax3.set_xlabel(r'Momentum $p$')
ax3.set_ylabel(r'Energy $E$')
ax3.set_title('Dirac Spectrum\nParticle & Antiparticle', color=BLUE)
ax3.legend(fontsize=9, labelcolor=FG); ax3.grid(True)
ax3.set_xlim(-4, 4); ax3.set_ylim(-5, 5)

plt.tight_layout(pad=1.2)
save('fig01_dirac_spectrum.png', fig)
```

## fig02 · Spinor Solutions & Helicity (§4.3)

**Left**: Large vs small components of $u(p,s)$ — small component $\sim|\mathbf{p}|/\sqrt{E+m}$ vanishes NR. **Centre**: Completeness $\sum_s u^s\bar u^s=\not{p}+m$, normalisation. **Right**: Helicity eigenstates $h=\hat{\mathbf{p}}\cdot\mathbf{S}=\pm1/2$.

```python
import numpy as np
import matplotlib.pyplot as plt
import os

BG='#1e1e2e'; FG='#cdd6f4'; EDGE='#45475a'
BLUE='#7ec8e3'; YELLOW='#f9c74f'; GREEN='#90ee90'
RED='#e07070'; PURPLE='#c792ea'; ORANGE='#ffb347'
plt.rcParams.update({
    'figure.facecolor':BG,'axes.facecolor':BG,'text.color':FG,
    'axes.labelcolor':FG,'xtick.color':FG,'ytick.color':FG,
    'axes.edgecolor':EDGE,'grid.color':EDGE,'grid.alpha':0.3,
    'font.size':12,'axes.titlesize':13,'axes.titleweight':'bold',
    'axes.titlecolor':BLUE,'legend.framealpha':0.3,'legend.edgecolor':EDGE,
})
OUTDIR = 'figures/ch04'; os.makedirs(OUTDIR, exist_ok=True)
def save(name, fig=None):
    f = fig or plt.gcf(); path = f'{OUTDIR}/{name}'
    f.savefig(path, dpi=180, bbox_inches='tight', facecolor=BG)
    print(f'  saved -> {path}'); plt.show()

# ── fig02: Spinor solutions u(p,s) and v(p,s) ─────────────────────────────
fig, axes = plt.subplots(1, 3, figsize=(14, 5))
fig.patch.set_facecolor(BG)

# Left: spinor components vs momentum (helicity)
ax = axes[0]
p_mag = np.linspace(0, 5, 200)
m = 1.0
E = np.sqrt(p_mag**2 + m**2)
# Helicity spinor: upper/lower components
upper = np.sqrt(E + m)     # ~ sqrt(2m) for NR, sqrt(E) for UR
lower = p_mag / np.sqrt(E + m)   # ~ p/sqrt(2m) NR, sqrt(E) UR (small component)
ax.plot(p_mag, upper, color=BLUE,   lw=2.2, label=r'Large comp. $\sim\sqrt{E+m}$')
ax.plot(p_mag, lower, color=RED,    lw=2.2, label=r'Small comp. $\sim|\mathbf{p}|/\sqrt{E+m}$')
ax.plot(p_mag, np.sqrt(E), color=GREEN, lw=1.8, ls='--', alpha=0.7,
        label=r'$\sqrt{E}$ (ultrarelativistic)')
ax.axvline(m, color=YELLOW, lw=1.2, ls=':', label='$|p|=m$')
ax.set_xlabel(r'$|\mathbf{p}|$'); ax.set_ylabel('Spinor component magnitude')
ax.set_title('Dirac Spinor $u(p,s)$\n' + r'Large vs small components', color=BLUE)
ax.legend(fontsize=8, labelcolor=FG); ax.grid(True); ax.set_xlim(0, 5)

# Centre: completeness relations (visual table)
ax2 = axes[1]
ax2.axis('off')
items = [
    (0.50, 0.91, 'Spinor Completeness', BLUE, 13, 'bold'),
    (0.50, 0.79, r'$\sum_s u^s(p)\bar u^s(p) = \not{p}+m$', GREEN, 11, 'normal'),
    (0.50, 0.66, r'$\sum_s v^s(p)\bar v^s(p) = \not{p}-m$', RED,   11, 'normal'),
    (0.50, 0.53, r'$\bar u = u^\dagger\gamma^0$', FG, 11, 'normal'),
    (0.50, 0.41, 'Normalisation:', ORANGE, 10, 'bold'),
    (0.50, 0.32, r'$\bar u^r(p)u^s(p)=2m\,\delta^{rs}$', YELLOW, 11, 'normal'),
    (0.50, 0.22, r'$\bar v^r(p)v^s(p)=-2m\,\delta^{rs}$', YELLOW, 11, 'normal'),
    (0.50, 0.11, r'$\bar u^r(p)v^s(p)=0$', FG, 10, 'normal'),
]
for (xp, yp, txt, col, fs, fw) in items:
    ax2.text(xp, yp, txt, ha='center', va='center', color=col,
             fontsize=fs, fontweight=fw, transform=ax2.transAxes)
rect = plt.Rectangle((0.03, 0.03), 0.94, 0.94, fill=True,
                      facecolor=YELLOW, alpha=0.05,
                      edgecolor=YELLOW, lw=1.5, transform=ax2.transAxes)
ax2.add_patch(rect)
ax2.set_title('Spinor Relations', color=BLUE)

# Right: spin-1/2 on Bloch sphere (helicity)
ax3 = axes[2]
theta_arr = np.linspace(0, np.pi, 200)
phi_arr   = np.linspace(0, 2*np.pi, 200)
# Helicity +1/2 vs -1/2 for a given momentum direction
p_arr = np.linspace(-1, 1, 100)
helicity_p = 0.5 * np.ones_like(p_arr)
helicity_n = -0.5 * np.ones_like(p_arr)
# Spin projections of u(p,+) along p-axis
cos_th = np.linspace(-1, 1, 200)
h_plus  = 0.5 * (1 + cos_th)
h_minus = 0.5 * (1 - cos_th)
ax3.fill_between(cos_th, h_plus,  color=BLUE,   alpha=0.30, label=r'Helicity $h=+1/2$')
ax3.fill_between(cos_th, h_minus, color=RED,    alpha=0.30, label=r'Helicity $h=-1/2$')
ax3.plot(cos_th, h_plus,  color=BLUE,   lw=2.2)
ax3.plot(cos_th, h_minus, color=RED,    lw=2.2)
ax3.axhline(0.5, color=YELLOW, lw=1.2, ls='--', alpha=0.6)
ax3.plot([0],[0.5],'o', color=GREEN, ms=10, zorder=5, label=r'Transverse $\cos\theta=0$')
ax3.set_xlabel(r'$\cos\theta$ (angle w.r.t. $\mathbf{p}$)')
ax3.set_ylabel('Probability')
ax3.set_title('Helicity Eigenstates\n' + r'$h=\hat{\mathbf{p}}\cdot\mathbf{S}$', color=BLUE)
ax3.legend(fontsize=9, labelcolor=FG); ax3.grid(True)
ax3.set_xlim(-1, 1); ax3.set_ylim(-0.05, 1.05)

plt.tight_layout(pad=1.2)
save('fig02_spinors.png', fig)
```

## fig03 · Fermi Statistics & Quantization (§4.4)

**Left**: Anticommutation relations — Pauli exclusion $n_\mathbf{p}\in\{0,1\}$. **Centre**: Dirac sea — filled negative-energy states; hole = antiparticle. **Right**: Mode expansion with $b^s_\mathbf{p}$ (fermion), $c^{s\dagger}_\mathbf{p}$ (antifermion).

```python
import numpy as np
import matplotlib.pyplot as plt
import os

BG='#1e1e2e'; FG='#cdd6f4'; EDGE='#45475a'
BLUE='#7ec8e3'; YELLOW='#f9c74f'; GREEN='#90ee90'
RED='#e07070'; PURPLE='#c792ea'; ORANGE='#ffb347'
plt.rcParams.update({
    'figure.facecolor':BG,'axes.facecolor':BG,'text.color':FG,
    'axes.labelcolor':FG,'xtick.color':FG,'ytick.color':FG,
    'axes.edgecolor':EDGE,'grid.color':EDGE,'grid.alpha':0.3,
    'font.size':12,'axes.titlesize':13,'axes.titleweight':'bold',
    'axes.titlecolor':BLUE,'legend.framealpha':0.3,'legend.edgecolor':EDGE,
})
OUTDIR = 'figures/ch04'; os.makedirs(OUTDIR, exist_ok=True)
def save(name, fig=None):
    f = fig or plt.gcf(); path = f'{OUTDIR}/{name}'
    f.savefig(path, dpi=180, bbox_inches='tight', facecolor=BG)
    print(f'  saved -> {path}'); plt.show()

# ── fig03: Quantization — Fermi statistics + Dirac sea ────────────────────
fig, axes = plt.subplots(1, 3, figsize=(14, 5))
fig.patch.set_facecolor(BG)

# Left: anticommutation vs commutation — Pauli exclusion
ax = axes[0]
ax.axis('off')
items = [
    (0.50, 0.91, 'Fermionic Quantization', BLUE, 13, 'bold'),
    (0.50, 0.80, r'Anticommutation relations:', GREEN, 11, 'bold'),
    (0.50, 0.70, r'$\{b^r_\mathbf{p},\,b^{s\dagger}_\mathbf{q}\}=(2\pi)^3\delta^{(3)}(\mathbf{p}-\mathbf{q})\delta^{rs}$',
     GREEN, 10, 'normal'),
    (0.50, 0.59, r'$\{b^r_\mathbf{p},\,b^s_\mathbf{q}\}=\{b^{r\dagger}_\mathbf{p},\,b^{s\dagger}_\mathbf{q}\}=0$',
     GREEN, 10, 'normal'),
    (0.50, 0.47, r'Pauli exclusion:', RED, 11, 'bold'),
    (0.50, 0.37, r'$(b^\dagger_\mathbf{p})^2=0 \;\Rightarrow\; n_\mathbf{p}\in\{0,1\}$',
     RED, 11, 'normal'),
    (0.50, 0.25, r'Compare bosons:', YELLOW, 10, 'bold'),
    (0.50, 0.15, r'$[a_\mathbf{p},a^\dagger_\mathbf{q}]=\delta^{(3)}(\mathbf{p}-\mathbf{q})$',
     YELLOW, 10, 'normal'),
    (0.50, 0.06, r'$n_\mathbf{p}\in\{0,1,2,\ldots\}$', YELLOW, 10, 'normal'),
]
for (xp, yp, txt, col, fs, fw) in items:
    ax.text(xp, yp, txt, ha='center', va='center', color=col,
            fontsize=fs, fontweight=fw, transform=ax.transAxes)
rect = plt.Rectangle((0.02, 0.02), 0.96, 0.96, fill=True,
                      facecolor=GREEN, alpha=0.05,
                      edgecolor=GREEN, lw=1.5, transform=ax.transAxes)
ax.add_patch(rect)
ax.set_title('Fermi Statistics\nAnticommutators', color=BLUE)

# Centre: Dirac sea picture
ax2 = axes[1]
p_vals = np.linspace(-3, 3, 7)
E_vals = np.sqrt(p_vals**2 + 1.0)
# Filled negative-energy states (Dirac sea)
for pv in p_vals:
    Ev = np.sqrt(pv**2 + 1.0)
    ax2.plot(pv, -Ev, 's', color=BLUE, ms=12, zorder=5)
    ax2.plot(pv,  Ev, 'o', color=FG,   ms=8, zorder=4, alpha=0.4)
# One hole = positron
ax2.plot(p_vals[3], -E_vals[3], 's', color=BG, ms=14, zorder=6)
ax2.plot(p_vals[3], -E_vals[3], 's', color=RED, ms=14, zorder=6,
         markerfacecolor='none', markeredgewidth=2, label='Hole = antiparticle')
# One particle above
ax2.plot(p_vals[4], E_vals[4], 'o', color=ORANGE, ms=12, zorder=6,
         label='Excited particle')
ax2.axhline(0, color=EDGE, lw=0.8, ls='--')
ax2.fill_between(np.linspace(-3.5,3.5,2), -5, -1, color=BLUE, alpha=0.08,
                 label='Dirac sea (filled)')
ax2.fill_between(np.linspace(-3.5,3.5,2),  1,  5, color=FG,   alpha=0.04,
                 label='Empty positive states')
p_c = np.linspace(-3.5, 3.5, 200)
ax2.plot(p_c,  np.sqrt(p_c**2+1), color=BLUE,   lw=1.5, ls='--', alpha=0.5)
ax2.plot(p_c, -np.sqrt(p_c**2+1), color=BLUE,   lw=1.5, ls='--', alpha=0.5)
ax2.set_xlabel(r'Momentum $p$'); ax2.set_ylabel(r'Energy $E$')
ax2.set_title('Dirac Sea Picture\nHole = antiparticle', color=BLUE)
ax2.legend(fontsize=8, labelcolor=FG); ax2.grid(True)
ax2.set_xlim(-3.5, 3.5); ax2.set_ylim(-4.5, 4.5)

# Right: mode expansion and Hamiltonian
ax3 = axes[2]
ax3.axis('off')
items3 = [
    (0.50, 0.91, 'Dirac Field Mode Expansion', BLUE, 12, 'bold'),
    (0.50, 0.80, r'$\hat\psi=\int\frac{d^3p}{(2\pi)^3}\frac{1}{\sqrt{2\omega_\mathbf{p}}}$',
     FG, 10, 'normal'),
    (0.50, 0.71, r'$\times\sum_s\left(b^s_\mathbf{p}u^s(p)e^{ip\cdot x}+c^{s\dagger}_\mathbf{p}v^s(p)e^{-ip\cdot x}\right)$',
     FG, 10, 'normal'),
    (0.50, 0.58, r'$b^s_\mathbf{p}$: destroys fermion (spin $s$)', GREEN, 10, 'normal'),
    (0.50, 0.49, r'$c^{s\dagger}_\mathbf{p}$: creates antifermion', RED,   10, 'normal'),
    (0.50, 0.37, r'Normal-ordered Hamiltonian:', ORANGE, 10, 'bold'),
    (0.50, 0.27, r'$:\hat H:=\int\frac{d^3p}{(2\pi)^3}\omega_\mathbf{p}$', ORANGE, 10, 'normal'),
    (0.50, 0.18, r'$\times\sum_s\left(b^{s\dagger}_\mathbf{p}b^s_\mathbf{p}+c^{s\dagger}_\mathbf{p}c^s_\mathbf{p}\right)$',
     ORANGE, 10, 'normal'),
    (0.50, 0.06, r'Both particles & antiparticles have $E>0$', YELLOW, 9, 'normal'),
]
for (xp, yp, txt, col, fs, fw) in items3:
    ax3.text(xp, yp, txt, ha='center', va='center', color=col,
             fontsize=fs, fontweight=fw, transform=ax3.transAxes)
rect3 = plt.Rectangle((0.02, 0.02), 0.96, 0.96, fill=True,
                       facecolor=ORANGE, alpha=0.04,
                       edgecolor=ORANGE, lw=1.5, transform=ax3.transAxes)
ax3.add_patch(rect3)
ax3.set_title('Mode Expansion & Hamiltonian', color=BLUE)

plt.tight_layout(pad=1.2)
save('fig03_quantization.png', fig)
```

## fig04 · Dirac Propagator & Discrete Symmetries (§4.5–4.6)

**Left**: Dirac propagator $S_F(p)=i(\not{p}+m)/(p^2-m^2+i\varepsilon)$ — poles at $p^0=\pm\omega_\mathbf{p}$. **Centre**: CPT transformation table for bilinears $\bar\psi\Gamma\psi$. **Right**: C (particle$\leftrightarrow$antiparticle) and P ($\mathbf{p}\to-\mathbf{p}$) on Dirac spectrum.

```python
import numpy as np
import matplotlib.pyplot as plt
import os

BG='#1e1e2e'; FG='#cdd6f4'; EDGE='#45475a'
BLUE='#7ec8e3'; YELLOW='#f9c74f'; GREEN='#90ee90'
RED='#e07070'; PURPLE='#c792ea'; ORANGE='#ffb347'
plt.rcParams.update({
    'figure.facecolor':BG,'axes.facecolor':BG,'text.color':FG,
    'axes.labelcolor':FG,'xtick.color':FG,'ytick.color':FG,
    'axes.edgecolor':EDGE,'grid.color':EDGE,'grid.alpha':0.3,
    'font.size':12,'axes.titlesize':13,'axes.titleweight':'bold',
    'axes.titlecolor':BLUE,'legend.framealpha':0.3,'legend.edgecolor':EDGE,
})
OUTDIR = 'figures/ch04'; os.makedirs(OUTDIR, exist_ok=True)
def save(name, fig=None):
    f = fig or plt.gcf(); path = f'{OUTDIR}/{name}'
    f.savefig(path, dpi=180, bbox_inches='tight', facecolor=BG)
    print(f'  saved -> {path}'); plt.show()

# ── fig04: Feynman propagator + discrete symmetries CPT ───────────────────
fig, axes = plt.subplots(1, 3, figsize=(14, 5))
fig.patch.set_facecolor(BG)

# Left: Dirac propagator spectral function
ax = axes[0]
p0_arr = np.linspace(-4, 4, 600)
m = 1.0; eps = 0.08
for k2, col, lab in [(0, BLUE,   r'$|\mathbf{k}|^2=0$'),
                     (1, GREEN,  r'$|\mathbf{k}|^2=1$'),
                     (3, ORANGE, r'$|\mathbf{k}|^2=3$')]:
    denom = (p0_arr**2 - k2 - m**2)**2 + eps**2
    spec = eps / (np.pi * denom)
    spec /= spec.max()
    ax.plot(p0_arr, spec, color=col, lw=2.0, label=lab)
ax.set_xlabel(r'$p^0$')
ax.set_ylabel('Spectral function (arb.)')
ax.set_title('Dirac Propagator\n' +
             r'$S_F(p)=\frac{i(\not{p}+m)}{p^2-m^2+i\varepsilon}$', color=BLUE)
ax.legend(fontsize=9, labelcolor=FG); ax.grid(True); ax.set_xlim(-4, 4)

# Centre: CPT symmetry table
ax2 = axes[1]
ax2.axis('off')
headers = ['', 'C', 'P', 'T', 'CPT']
rows = [
    [r'$\psi$',          r'$-i\gamma^2\psi^*$', r'$\gamma^0\psi$',
     r'$i\gamma^1\gamma^3\psi^*$', r'$i\gamma^5\psi$'],
    [r'$\bar\psi\psi$',  r'$+$', r'$+$', r'$+$', r'$+$'],
    [r'$\bar\psi\gamma^\mu\psi$', r'$-$', r'$(-1)^\mu$', r'$(-1)^\mu$', r'$-$'],
    [r'$\bar\psi\gamma^5\psi$',  r'$-$', r'$-$', r'$+$', r'$+$'],
    [r'$j^\mu_{\rm EM}$', r'$-j^\mu$', r'$j_\mu$', r'$j_\mu$', r'$-j^\mu$'],
]
col_colors = [FG, GREEN, BLUE, YELLOW, RED]
ax2.text(0.50, 0.96, 'Discrete Symmetries C, P, T', ha='center',
         color=BLUE, fontsize=12, fontweight='bold', transform=ax2.transAxes)
xs = [0.12, 0.30, 0.48, 0.66, 0.84]
for xi, (hd, cc) in zip(xs, zip(headers, col_colors)):
    ax2.text(xi, 0.88, hd, ha='center', color=cc, fontsize=10,
             fontweight='bold', transform=ax2.transAxes)
ax2.plot([0.02, 0.98], [0.84, 0.84], color=EDGE, lw=0.8, transform=ax2.transAxes)
for ri, row in enumerate(rows):
    yp = 0.74 - ri*0.14
    for xi, (val, cc) in zip(xs, zip(row, col_colors)):
        ax2.text(xi, yp, val, ha='center', va='center', color=cc,
                 fontsize=9, transform=ax2.transAxes)
    if ri < len(rows)-1:
        ax2.plot([0.02, 0.98], [yp-0.07, yp-0.07],
                 color=EDGE, lw=0.5, transform=ax2.transAxes)
rect2 = plt.Rectangle((0.02, 0.02), 0.96, 0.96, fill=True,
                       facecolor=PURPLE, alpha=0.05,
                       edgecolor=PURPLE, lw=1.2, transform=ax2.transAxes)
ax2.add_patch(rect2)
ax2.set_title('CPT Transformation Table', color=BLUE)

# Right: charge conjugation — particle <-> antiparticle
ax3 = axes[2]
p_c = np.linspace(-3.5, 3.5, 300)
E_c = np.sqrt(p_c**2 + 1.0)
ax3.plot(p_c,  E_c, color=BLUE, lw=2.5, label=r'Fermion $e^-$')
ax3.plot(p_c, -E_c, color=RED,  lw=2.5, label=r'Antifermion $e^+$ (C)')
# P: p -> -p (reflection)
ax3.annotate('', xy=(-2, np.sqrt(4+1)), xytext=(2, np.sqrt(4+1)),
             arrowprops=dict(arrowstyle='<->', color=GREEN, lw=2.0, mutation_scale=14))
ax3.text(0, np.sqrt(5)+0.3, r'$P$: $\mathbf{p}\to-\mathbf{p}$',
         ha='center', color=GREEN, fontsize=10)
# C: fermion -> antifermion
ax3.annotate('', xy=(1.5, -np.sqrt(1.5**2+1)-0.1),
             xytext=(1.5, np.sqrt(1.5**2+1)+0.1),
             arrowprops=dict(arrowstyle='<->', color=ORANGE, lw=2.0, mutation_scale=14))
ax3.text(2.2, 0, r'$C$', ha='center', color=ORANGE, fontsize=12, fontweight='bold')
ax3.axhline(0, color=EDGE, lw=0.8, ls='--')
ax3.set_xlabel(r'Momentum $\mathbf{p}$')
ax3.set_ylabel(r'Energy $E$')
ax3.set_title('C and P Transformations\n' + r'on Dirac spinors', color=BLUE)
ax3.legend(fontsize=9, labelcolor=FG); ax3.grid(True)
ax3.set_xlim(-3.5, 3.5); ax3.set_ylim(-4.5, 4.5)

plt.tight_layout(pad=1.2)
save('fig04_propagator_cpt.png', fig)
```
