## Visualization 2 · QED: Cross Section & Running $\alpha$ (Ch 2)

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
fig.patch.set_facecolor(BG)

# ── Left: sigma(e+e-→μ+μ-) vs sqrt(s) ────────────────────────────────────
ax = axes[0]
ax.set_facecolor(BG)

sqrt_s = np.logspace(np.log10(0.2), np.log10(200), 600)
alpha0 = 1/137.036
conv = 3.89e5  # nb per GeV^-2

sigma_qed = 4*np.pi*alpha0**2/(3*sqrt_s**2)*conv  # nb

# Z resonance (Breit-Wigner)
mZ, GZ = 91.2, 2.495
BW = (mZ*GZ)**2/((sqrt_s**2-mZ**2)**2+(mZ*GZ)**2)
sigma_total = sigma_qed*(1 + 400*BW)

ax.loglog(sqrt_s, sigma_qed, color=TEAL, lw=2.5, label=r'$\sigma_0=4\pi\alpha^2/3s$')
ax.loglog(sqrt_s, sigma_total, color=RED, lw=2, ls='--', label=r'With $Z^0$ resonance')

ax.axvline(mZ, color=RED, lw=1.5, ls=':', alpha=0.7)
ax.text(mZ*1.05, 3e4, r'$m_Z=91.2$ GeV', fontsize=9, color=RED)
ax.text(5, 100, r'$\sigma\propto 1/s$', color=TEAL, fontsize=10, rotation=-22)

ax.set_xlabel(r'$\sqrt{s}$ (GeV)')
ax.set_ylabel(r'$\sigma$ (nb)')
ax.set_title(r'$e^+e^-\to\mu^+\mu^-$'+" Cross Section\n"+r'$\propto 1/s$ (QED) with $Z^0$ peak')
ax.legend(fontsize=9, labelcolor=FG)
ax.set_xlim(0.2, 200)
ax.grid(True, which='both', alpha=0.3)

# ── Middle: Running alpha_em ───────────────────────────────────────────────
ax2 = axes[1]
ax2.set_facecolor(BG)

q_arr = np.logspace(-3, 2.3, 400)  # GeV
alpha0_inv = 137.036

def beta_f(q, mf, Qf, nc=1):
    return np.where(q > mf, nc*Qf**2/(3*np.pi)*np.log(q/mf), 0)

di = np.zeros_like(q_arr)
di += beta_f(q_arr, 0.000511, 1, 1)   # e
di += beta_f(q_arr, 0.1057, 1, 1)     # mu
di += beta_f(q_arr, 1.777, 1, 1)      # tau
di += beta_f(q_arr, 0.003, 2/3, 3)    # u
di += beta_f(q_arr, 0.005, 1/3, 3)    # d
di += beta_f(q_arr, 0.10, 1/3, 3)     # s
di += beta_f(q_arr, 1.28, 2/3, 3)     # c
di += beta_f(q_arr, 4.18, 1/3, 3)     # b

alpha_inv = alpha0_inv - di

ax2.semilogx(q_arr, alpha_inv, color=TEAL, lw=2.5, label=r'$1/\alpha_{\rm em}(q)$')
ax2.axhline(alpha0_inv, color=EDGE, lw=1.5, ls=':', alpha=0.7, label=r'$1/\alpha_0=137$')

aZ_inv = np.interp(91.2, q_arr, alpha_inv)
ax2.axvline(91.2, color=RED, lw=1.5, ls=':', alpha=0.8)
ax2.plot(91.2, aZ_inv, 'D', color=RED, ms=12, zorder=5)
ax2.text(95, aZ_inv+0.8, fr'$1/\alpha(m_Z)\approx{aZ_inv:.0f}$', fontsize=9, color=RED)

ax2.text(0.002, 130, 'Anti-screening:\n'+r'$\alpha_{\rm em}$ grows at high $E$', fontsize=9, color=TEAL,
         bbox=dict(boxstyle='round,pad=0.3', facecolor=BG, edgecolor=TEAL, alpha=0.8))

ax2.set_xlabel(r'Energy scale $q$ (GeV)')
ax2.set_ylabel(r'$1/\alpha_{\rm em}(q)$')
ax2.set_title(r'Running of $\alpha_{\rm em}$'+"\n"+r'Anti-screening: opposite to QCD!')
ax2.legend(fontsize=9, labelcolor=FG)
ax2.set_xlim(0.001, 200)
ax2.set_ylim(125, 138)
ax2.grid(True, which='both', alpha=0.3)

# ── Right: Feynman vertex schematic ──────────────────────────────────────
ax3 = axes[2]
ax3.set_facecolor(BG); ax3.axis('off')
ax3.set_xlim(0, 1); ax3.set_ylim(0, 1)

def wavy(ax, x1, y1, x2, y2, col=TEAL, nw=5, amp=0.025):
    t = np.linspace(0, 1, 200)
    dx, dy = x2-x1, y2-y1
    L = np.sqrt(dx**2+dy**2)
    px, py = -dy/L, dx/L
    wave = amp*np.sin(nw*2*np.pi*t)
    ax.plot(x1+t*dx+wave*px, y1+t*dy+wave*py, color=col, lw=2)

# QED vertex
ax3.text(0.5, 0.95, 'QED Vertex', ha='center', fontsize=13, color=TEAL, fontweight='bold')
ax3.plot(0.5, 0.72, 'o', color=TEAL, ms=12, zorder=5)
ax3.annotate('', xy=(0.5, 0.72), xytext=(0.20, 0.88),
             arrowprops=dict(arrowstyle='->', color=BLUE, lw=2.5, mutation_scale=15))
ax3.text(0.12, 0.90, r'$e^-$', fontsize=12, color=BLUE)
ax3.annotate('', xy=(0.80, 0.88), xytext=(0.5, 0.72),
             arrowprops=dict(arrowstyle='->', color=BLUE, lw=2.5, mutation_scale=15))
ax3.text(0.82, 0.90, r"$e^-$", fontsize=12, color=BLUE)
wavy(ax3, 0.5, 0.72, 0.5, 0.48, col=TEAL, nw=4)
ax3.text(0.56, 0.57, r'$\gamma$', fontsize=12, color=TEAL)

ax3.text(0.5, 0.42, r'Vertex factor: $-ie\gamma^\mu$', ha='center', fontsize=11, color=YELL,
         bbox=dict(boxstyle='round,pad=0.3', facecolor=BG, edgecolor=YELL, alpha=0.8))

# g-2 loop
ax3.text(0.5, 0.30, r'One-loop $g-2$:', ha='center', fontsize=10, color=FG)
ax3.text(0.5, 0.22, r'$a_e=\frac{\alpha}{2\pi}+\ldots$', ha='center', fontsize=12, color=GREEN,
         bbox=dict(boxstyle='round,pad=0.3', facecolor=BG, edgecolor=GREEN, alpha=0.8))
ax3.text(0.5, 0.11, r'Theory vs exp: 12 s.f. agreement', ha='center', fontsize=9, color=PURP)
ax3.text(0.5, 0.03, r'Most precise test of any theory!', ha='center', fontsize=9, color=PURP, fontstyle='italic')

ax3.set_title('QED Building Blocks\nVertex + Anomalous $g-2$', color=TEAL)

plt.tight_layout(pad=1.5)
save('ch02_qed.png')
```

## Visualization 3 · QCD: $\alpha_s$, $R$-Ratio, Potential (Ch 3)

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
fig.patch.set_facecolor(BG)

# ── Left: alpha_s(mu) ─────────────────────────────────────────────────────
ax = axes[0]
ax.set_facecolor(BG)

Lambda_QCD = 0.2
mu_arr = np.logspace(0, 3, 400)  # 1 to 1000 GeV
b0 = 7.0
alpha_s_run = 2*np.pi/(b0*np.log(mu_arr/Lambda_QCD))
alpha_s_run = np.clip(alpha_s_run, 0, 1)

ax.semilogx(mu_arr, alpha_s_run, color=YELL, lw=2.5, label=r'$\alpha_s(\mu)$ (1-loop)')
ax.fill_between(mu_arr, 0, alpha_s_run, color=YELL, alpha=0.1)

# PDG data
data_mu  = [1.5, 5.0, 10.0, 31.6, 91.2, 189.0]
data_as  = [0.33, 0.21, 0.18, 0.147, 0.118, 0.11]
data_err = [0.03, 0.02, 0.015, 0.005, 0.002, 0.005]
ax.errorbar(data_mu, data_as, yerr=data_err, fmt='D', color=TEAL,
            ms=9, capsize=4, elinewidth=1.5, label='PDG data', zorder=5)

ax.axvline(91.2, color=RED, lw=1.5, ls=':', alpha=0.8)
ax.text(95, 0.5, r'$m_Z$'+"\n"+r'$\alpha_s=0.118$', fontsize=8.5, color=RED)
ax.axvline(Lambda_QCD, color=PURP, lw=1.5, ls=':', alpha=0.7)
ax.text(0.22, 0.32, r'$\Lambda_{\rm QCD}$', fontsize=8.5, color=PURP)

ax.text(100, 0.55, r'$\beta<0$: AF!', fontsize=12, color=YELL, fontweight='bold')
ax.set_xlabel(r'Energy scale $\mu$ (GeV)')
ax.set_ylabel(r'$\alpha_s(\mu)$')
ax.set_title('QCD Asymptotic Freedom\n'+r'$\alpha_s\to 0$ as $\mu\to\infty$')
ax.legend(fontsize=9, labelcolor=FG)
ax.set_xlim(1, 1000); ax.set_ylim(0, 0.65)
ax.grid(True, which='both', alpha=0.3)

# ── Middle: R-ratio ───────────────────────────────────────────────────────
ax2 = axes[1]
ax2.set_facecolor(BG)

sqrt_s_R = np.linspace(0.5, 15, 500)
Nc = 3
R_uds  = Nc*(4/9+1/9+1/9)    # = 2
R_udsc = Nc*(4/9+1/9+1/9+4/9)  # = 10/3
R_udscb= Nc*(4/9+1/9+1/9+4/9+1/9)  # = 11/3

R_th = np.where(sqrt_s_R < 3.1, R_uds,
       np.where(sqrt_s_R < 9.5, R_udsc, R_udscb))

ax2.step(sqrt_s_R, R_th, color=TEAL, lw=2.5, label=r'$R=N_c\sum Q_f^2$ (theory)')

# Resonances
for (E, Rp, lbl, col) in [(3.097, 8.5, r'$J/\psi$ (charm)', YELL),
                            (9.460, 14, r'$\Upsilon$ (bottom)', GREEN)]:
    ax2.annotate('', xy=(E, Rp), xytext=(E, R_udsc if E<5 else R_udscb),
                arrowprops=dict(arrowstyle='->', color=col, lw=2.5))
    ax2.text(E+0.1, Rp*0.9, lbl, fontsize=8.5, color=col)

for R_val, lbl, col in [(R_uds, r'$R=2$ (uds)', BLUE),
                         (R_udsc, r'$R=10/3$ (udsc)', ORAN),
                         (R_udscb, r'$R=11/3$ (udscb)', RED)]:
    ax2.axhline(R_val, color=col, lw=1.2, ls=':', alpha=0.7)
    ax2.text(0.6, R_val+0.1, f'R={R_val:.2f}', fontsize=8, color=col)

ax2.set_xlabel(r'$\sqrt{s}$ (GeV)')
ax2.set_ylabel(r'$R$ ratio')
ax2.set_title(r'$R$-Ratio: Evidence for $N_c=3$'+"\nStep jumps at quark thresholds")
ax2.legend(fontsize=9, labelcolor=FG)
ax2.set_xlim(0.5, 15); ax2.set_ylim(0, 16)
ax2.grid(True, alpha=0.3)

# ── Right: V(r) potential ──────────────────────────────────────────────────
ax3 = axes[2]
ax3.set_facecolor(BG)

r_fm = np.linspace(0.05, 2.5, 400)
r_gev = r_fm * 5.068  # fm → GeV^{-1}
alpha_s_v = 0.30
sigma_v = 0.18  # GeV^2

V_coulomb = -4*alpha_s_v/(3*r_gev) * 1000  # MeV
V_linear  =  sigma_v * r_gev * 1000         # MeV
V_total   = V_coulomb + V_linear

ax3.plot(r_fm, V_total, color=TEAL, lw=2.5, label=r'$V=V_{\rm Coul}+\sigma r$')
ax3.plot(r_fm, V_coulomb, color=BLUE, lw=1.8, ls='--', alpha=0.7, label=r'Coulomb: $-4\alpha_s/3r$')
ax3.plot(r_fm, V_linear, color=RED, lw=1.8, ls='--', alpha=0.7, label=r'Linear: $\sigma r$')

# String breaking
V_break = 270  # MeV
ax3.axhline(V_break, color=YELL, lw=1.5, ls=':', alpha=0.8)
ax3.text(0.15, V_break+25, r'String breaks: $2m_\pi\approx 270$ MeV', fontsize=8.5, color=YELL)

ax3.axhline(0, color=EDGE, lw=1)
ax3.set_xlabel('Separation $r$ (fm)')
ax3.set_ylabel('$V(r)$ (MeV)')
ax3.set_title('Cornell Potential\n'+r'Coulomb (short-range) + Linear (confinement)')
ax3.legend(fontsize=9, labelcolor=FG)
ax3.set_xlim(0.05, 2.5); ax3.set_ylim(-700, 900)
ax3.grid(True, alpha=0.3)

plt.tight_layout(pad=1.5)
save('ch03_qcd.png')
```

## Visualization 4 · Higgs Mechanism & Neutrino Oscillations (Ch 4)

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
fig.patch.set_facecolor(BG)

# ── Left: Higgs potential ──────────────────────────────────────────────────
ax = axes[0]
ax.set_facecolor(BG)

phi = np.linspace(-2, 2, 400)
mu2, lam = 1.0, 0.5
V = -mu2*phi**2 + lam*phi**4
V_no = mu2*phi**2 + lam*phi**4  # unbroken

ax.plot(phi, V, color=TEAL, lw=2.5, label=r'$V=-\mu^2|\phi|^2+\lambda|\phi|^4$ (SSB)')
ax.plot(phi, V_no, color=EDGE, lw=2, ls='--', alpha=0.7, label=r'$\mu^2<0$: no SSB')

v_min = np.sqrt(mu2/(2*lam))
V_min = -mu2*v_min**2 + lam*v_min**4
ax.plot([-v_min, v_min], [V_min, V_min], 'D', color=YELL, ms=14, zorder=5,
        label=r'VEV: $\phi=\pm v/\sqrt{2}$, $v=246$ GeV')

# Higgs mass arrow
x_h = v_min + 0.35
V_h = -mu2*x_h**2 + lam*x_h**4
ax.annotate('', xy=(x_h, V_h), xytext=(v_min, V_min),
            arrowprops=dict(arrowstyle='<->', color=RED, lw=2))
ax.text(x_h+0.08, (V_min+V_h)/2, r'$m_H=125$ GeV', fontsize=9, color=RED)

# Goldstone direction
ax.annotate('', xy=(-v_min, V_min+0.05), xytext=(v_min, V_min+0.05),
            arrowprops=dict(arrowstyle='<->', color=GREEN, lw=2))
ax.text(0, V_min+0.18, r'Goldstone $\to W^\pm,Z$ masses', ha='center', fontsize=9, color=GREEN)

ax.axhline(0, color=EDGE, lw=1)
ax.set_xlabel(r'$\mathrm{Re}(\phi)$')
ax.set_ylabel(r'$V(\phi)$')
ax.set_title('Higgs Potential\n'+r'"Mexican Hat" — SSB gives $W,Z$ mass')
ax.legend(fontsize=8.5, labelcolor=FG)
ax.set_ylim(-0.6, 1.5)
ax.grid(True, alpha=0.3)

# ── Middle: W/Z mass bar chart ─────────────────────────────────────────────
ax2 = axes[1]
ax2.set_facecolor(BG)

bosons_m = [('$\\gamma$', 0, YELL), ('$W^\\pm$', 80.4, TEAL), ('$Z^0$', 91.2, BLUE), ('$H$', 125.25, PURP)]
xpos = range(len(bosons_m))
masses_b = [m for _, m, _ in bosons_m]
colors_b = [c for _, _, c in bosons_m]
labels_b = [l for l, _, _ in bosons_m]

bars = ax2.bar(xpos, masses_b, color=colors_b, alpha=0.85, edgecolor=EDGE, lw=1.5, width=0.6)

for x, m, col in zip(xpos, masses_b, colors_b):
    if m > 0:
        ax2.text(x, m+2, f'{m} GeV', ha='center', fontsize=10, color=col, fontweight='bold')
    else:
        ax2.text(x, 4, 'massless', ha='center', fontsize=9, color=col)

ax2.set_xticks(list(xpos))
ax2.set_xticklabels(labels_b, fontsize=13)
ax2.set_ylabel('Mass (GeV)')
ax2.set_title('Electroweak Boson Masses\n'+r'$m_W=gv/2$, $m_Z=\sqrt{g^2+g^{\prime 2}}v/2$, $m_\gamma=0$')

# Annotations
ax2.text(2.5, 50, r'$m_W/m_Z=\cos\theta_W$', fontsize=9.5, color=TEAL,
         bbox=dict(boxstyle='round,pad=0.3', facecolor=BG, edgecolor=TEAL, alpha=0.8))
ax2.text(2.5, 30, r'$v=246$ GeV (Higgs VEV)', fontsize=9, color=YELL)
ax2.set_ylim(0, 150)
ax2.grid(True, axis='y', alpha=0.3)

# ── Right: Neutrino oscillations ──────────────────────────────────────────
ax3 = axes[2]
ax3.set_facecolor(BG)

L_over_E = np.linspace(0, 2000, 800)  # km/GeV
theta23 = np.radians(45)
Dm2_atm = 2.5e-3  # eV^2

P_survival = 1 - np.sin(2*theta23)**2 * np.sin(1.27*Dm2_atm*L_over_E)**2
P_appear   = 1 - P_survival

ax3.plot(L_over_E, P_survival, color=TEAL, lw=2.5, label=r'$P(\nu_\mu\to\nu_\mu)$')
ax3.plot(L_over_E, P_appear,   color=RED,  lw=2.5, ls='--', label=r'$P(\nu_\mu\to\nu_\tau)$')
ax3.fill_between(L_over_E, 0, P_survival, color=TEAL, alpha=0.1)

# Experiment markers
ax3.axvline(295/0.6, color=GREEN, lw=2, ls=':', alpha=0.9)
ax3.text(295/0.6+20, 0.85, 'T2K\n(L/E≈490)', fontsize=8.5, color=GREEN)
ax3.axvline(730/3.0, color=YELL, lw=2, ls=':', alpha=0.9)
ax3.text(730/3.0+15, 0.08, 'NOvA\n(L/E≈243)', fontsize=8.5, color=YELL)
ax3.axvline(500, color=ORAN, lw=1.5, ls=':', alpha=0.8)
ax3.text(520, 0.50, 'Super-K\n(atm)', fontsize=8.5, color=ORAN)

ax3.text(1400, 0.82,
         r'$\sin^2(2\theta_{23})\approx 1$'+"\n"+
         r'$\Delta m^2_{32}=2.5\times10^{-3}$ eV$^2$',
         fontsize=8.5, color=TEAL,
         bbox=dict(boxstyle='round,pad=0.3', facecolor=BG, edgecolor=TEAL, alpha=0.8))

ax3.set_xlabel(r'$L/E$ (km/GeV)')
ax3.set_ylabel('Oscillation probability')
ax3.set_title(r'Neutrino Oscillation $\nu_\mu\to\nu_\tau$'+"\n"+
              r'Nobel Prize 2015 (Kajita & McDonald)')
ax3.legend(fontsize=9, labelcolor=FG)
ax3.set_xlim(0, 2000); ax3.set_ylim(-0.05, 1.1)
ax3.grid(True, alpha=0.3)

plt.tight_layout(pad=1.5)
save('ch04_weak.png')
```

## Visualization 5 · BSM: Hierarchy Problem, Rotation Curves, Dark Matter (Ch 5)

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
fig.patch.set_facecolor(BG)

# ── Left: Hierarchy problem scale comparison ──────────────────────────────
ax = axes[0]
ax.set_facecolor(BG)

scales_hp = [
    (r'$m_H = 125$ GeV', np.log10(125), TEAL),
    (r'$m_t = 173$ GeV', np.log10(173), YELL),
    (r'$m_Z = 91$ GeV', np.log10(91.2), BLUE),
    (r'$v = 246$ GeV', np.log10(246), GREEN),
    (r'$M_{\rm GUT} \sim 10^{16}$ GeV', 16, PURP),
    (r'$M_{\rm Pl} = 2.4\times 10^{18}$ GeV', np.log10(2.4e18), RED),
]

for i, (lbl, logE, col) in enumerate(scales_hp):
    ax.barh(i, logE, color=col, alpha=0.8, edgecolor=EDGE, lw=1.2, height=0.6)
    ax.text(logE+0.3, i, lbl, va='center', fontsize=9.5, color=col)

# Hierarchy arrow
ax.annotate('', xy=(np.log10(2.4e18), 0), xytext=(np.log10(2.4e18), 3),
            arrowprops=dict(arrowstyle='<->', color=RED, lw=2.5))
ax.text(np.log10(2.4e18)+0.3, 1.5, r'34 orders!', fontsize=10, color=RED,
        va='center', fontweight='bold')

ax.set_yticks([])
ax.set_xlabel(r'$\log_{10}(E/\mathrm{GeV})$')
ax.set_title('Hierarchy Problem\n'+r'$m_H\ll M_{\rm Pl}$: 34 orders of magnitude!')
ax.set_xlim(0, 22)
ax.grid(True, axis='x', alpha=0.3)

# ── Middle: Galaxy rotation curve ─────────────────────────────────────────
ax2 = axes[1]
ax2.set_facecolor(BG)

r_kpc = np.linspace(0.1, 25, 400)
v_bulge = 200 * np.where(r_kpc < 2, r_kpc/2, 2/np.sqrt(r_kpc/2))  # schematic
v_disk  = 150 * np.where(r_kpc < 6, r_kpc/6, np.sqrt(6/r_kpc))
v_halo  = 120 * np.ones_like(r_kpc)  # flat from DM halo

# Total (visible only)
v_visible = np.sqrt(v_bulge**2/2 + v_disk**2)
# With DM
v_total   = np.sqrt(v_bulge**2/2 + v_disk**2 + v_halo**2*0.8*(1-np.exp(-r_kpc/5)))

ax2.plot(r_kpc, v_visible, color=YELL, lw=2.5, ls='--', label='Visible matter only')
ax2.plot(r_kpc, v_total, color=TEAL, lw=2.5, label='Observed (flat)')

# Scatter plot (simulated data)
np.random.seed(42)
r_data = np.array([0.5,1,2,4,6,8,10,13,17,21,25])
v_data = np.interp(r_data, r_kpc, v_total) + np.random.normal(0, 8, len(r_data))
ax2.scatter(r_data, v_data, color=TEAL, s=70, zorder=5, alpha=0.9)

# DM halo annotation
ax2.fill_between(r_kpc, v_visible, v_total, color=RED, alpha=0.15, label='Dark matter halo')
ax2.text(14, 100, r'DM halo: $\rho\propto 1/r^2$', fontsize=9, color=RED,
         bbox=dict(boxstyle='round,pad=0.3', facecolor=BG, edgecolor=RED, alpha=0.8))

ax2.set_xlabel('Galactic radius $r$ (kpc)')
ax2.set_ylabel('Rotation velocity $v$ (km/s)')
ax2.set_title('Galaxy Rotation Curve\n'+r'Flat $v(r)$ requires dark matter halo')
ax2.legend(fontsize=9, labelcolor=FG)
ax2.set_xlim(0, 25); ax2.set_ylim(0, 280)
ax2.grid(True, alpha=0.3)

# ── Right: Universe energy budget (pie) ────────────────────────────────────
ax3 = axes[2]
ax3.set_facecolor(BG)

sizes  = [68.3, 26.8, 4.9]
labels_pie = ['Dark Energy\n68.3%\n(cosmological constant?)', 
              'Dark Matter\n26.8%\n(WIMP? axion?)',
              'Ordinary\nMatter 4.9%']
colors_pie = [PURP, TEAL, YELL]
explode    = [0.03, 0.03, 0.08]

wedges, texts = ax3.pie(sizes, explode=explode, labels=labels_pie,
                         colors=colors_pie, startangle=90, autopct='',
                         textprops={'color':FG,'fontsize':9.5},
                         wedgeprops={'linewidth':1.5,'edgecolor':BG})

# Sub-breakdown of ordinary matter
ax3.text(0, -1.55,
         r'Ordinary matter breakdown:'+"\n"+
         r'Stars $\approx 0.5\%$ | Gas $\approx 4\%$ | Neutrinos $<0.01\%$',
         ha='center', fontsize=8.5, color=YELL)

ax3.set_title('Universe Energy Budget (Planck 2018)\n'+
              r'$\Omega_{\rm DE}=0.683,\;\Omega_{\rm DM}=0.268,\;\Omega_b=0.049$', color=TEAL)

plt.tight_layout(pad=1.5)
save('ch05_bsm.png')
```
