## Fig 01 — Fine Structure Diagram

```python

# Fig 01: Hydrogen energy levels — gross, fine, Lamb, hyperfine
# Shows n=1 and n=2 with all corrections (not to scale)

fig, ax = plt.subplots(figsize=(10, 7))
ax.set_xlim(0, 10)
ax.set_ylim(-0.5, 8.5)
ax.axis('off')

def draw_level(ax, x1, x2, y, color, lw=2.0, ls='-'):
    ax.hlines(y, x1, x2, colors=color, linewidths=lw, linestyles=ls)

def label_right(ax, x, y, text, color, fontsize=9):
    ax.text(x+0.08, y, text, va='center', color=color, fontsize=fontsize)

def label_left(ax, x, y, text, color, fontsize=9):
    ax.text(x-0.08, y, text, va='center', ha='right', color=color, fontsize=fontsize)

# ── Gross structure ────────────────────────────────────────
# n=1
draw_level(ax, 0.3, 1.5, 1.0, BLUE, lw=3.)
label_left(ax, 0.3, 1.0, '$n=1$', BLUE, 11)

# n=2
draw_level(ax, 0.3, 1.5, 5.5, BLUE, lw=3.)
label_left(ax, 0.3, 5.5, '$n=2$', BLUE, 11)

ax.text(0.9, -0.3, 'Gross\nstructure', ha='center', color=BLUE, fontsize=9)

# ── Fine structure (n=2) ───────────────────────────────────
# j=1/2: 2s_{1/2} and 2p_{1/2} (degenerate in Dirac)
# j=3/2: 2p_{3/2} (higher)
y_j12 = 5.3
y_j32 = 6.1

draw_level(ax, 2.2, 3.8, y_j12, GREEN, lw=2.5)
draw_level(ax, 2.2, 3.8, y_j32, GREEN, lw=2.5)

# n=1: j=1/2 only
draw_level(ax, 2.2, 3.8, 1.0, GREEN, lw=2.5)

# Brace / bracket
ax.annotate('', xy=(1.7, y_j12), xytext=(1.7, y_j32),
            arrowprops=dict(arrowstyle='<->', color=GREEN, lw=1.2))
ax.text(1.55, (y_j12+y_j32)/2, r'$\Delta E_{\rm fs}$',
        ha='right', va='center', color=GREEN, fontsize=8)

label_right(ax, 3.8, y_j12, r'$2s_{1/2},\,2p_{1/2}$  ($j=\frac{1}{2}$)', GREEN, 9)
label_right(ax, 3.8, y_j32, r'$2p_{3/2}$  ($j=\frac{3}{2}$)', GREEN, 9)
label_right(ax, 3.8, 1.0,  r'$1s_{1/2}$  ($j=\frac{1}{2}$)', GREEN, 9)

ax.text(3.0, 4.6, 'Fine\nstructure', ha='center', color=GREEN, fontsize=9)

# ── Lamb shift (2s_{1/2} shifted up relative to 2p_{1/2}) ──
y_2s = y_j12 + 0.38
y_2p12 = y_j12

draw_level(ax, 5.0, 6.5, y_2s,  YELLOW, lw=2.5)
draw_level(ax, 5.0, 6.5, y_2p12, YELLOW, lw=2.5)
draw_level(ax, 5.0, 6.5, y_j32,  YELLOW, lw=2.5)
draw_level(ax, 5.0, 6.5, 1.0,   YELLOW, lw=2.5)

ax.annotate('', xy=(5.05, y_2p12), xytext=(5.05, y_2s),
            arrowprops=dict(arrowstyle='<->', color=YELLOW, lw=1.2))
ax.text(5.45, (y_2s+y_2p12)/2,
        '1057\nMHz', ha='right', va='center', color=YELLOW, fontsize=7.5)

label_right(ax, 6.5, y_2s,   r'$2s_{1/2}$', YELLOW, 9)
label_right(ax, 6.5, y_2p12, r'$2p_{1/2}$', YELLOW, 9)
label_right(ax, 6.5, y_j32,  r'$2p_{3/2}$', YELLOW, 9)

ax.text(5.75, 4.6, 'Lamb\nshift\n(QED)', ha='center', color=YELLOW, fontsize=9)

# ── Hyperfine (n=1: F=0, F=1) ──────────────────────────────
y_F1 = 1.35
y_F0 = 0.65

draw_level(ax, 7.8, 9.2, y_F1, RED, lw=2.5)
draw_level(ax, 7.8, 9.2, y_F0, RED, lw=2.5)
# n=2 levels (just show as degenerate for simplicity)
for yy in [y_2s+0.1, y_2p12+0.1, y_j32+0.1]:
    draw_level(ax, 7.8, 9.2, yy, RED, lw=1.5, ls='--')

ax.annotate('', xy=(7.6, y_F0), xytext=(7.6, y_F1),
            arrowprops=dict(arrowstyle='<->', color=RED, lw=1.2))
ax.text(7.45, (y_F0+y_F1)/2, '21 cm\n1420 MHz',
        ha='right', va='center', color=RED, fontsize=7.5)

label_right(ax, 9.2, y_F1, r'$F=1$', RED, 9)
label_right(ax, 9.2, y_F0, r'$F=0$', RED, 9)

ax.text(8.5, 4.6, 'Hyperfine\n(nuclear spin)', ha='center', color=RED, fontsize=9)

# Column headers
for x, lbl, col in [(0.9,'Gross',BLUE),(3.0,'Fine\nstructure',GREEN),
                     (5.75,'+ Lamb',YELLOW),(8.5,'+ Hyperfine',RED)]:
    ax.text(x, 8.2, lbl, ha='center', color=col, fontsize=10, fontweight='bold')

ax.set_title('Hydrogen Energy Levels: All Corrections', color='#cdd6f4', pad=20)
plt.tight_layout()
SAVE('fig01_fine_structure.png'); plt.show(); print('Fig 01 saved.')
```

## Fig 02 — Shell Filling

```python

# Fig 02: Atomic orbital energy ordering and Madelung rule
# Subshells ordered by energy with capacities shown

fig, ax = plt.subplots(figsize=(9, 6))
ax.set_xlim(-1, 9); ax.set_ylim(-0.5, 10.5)
ax.axis('off')

# Subshells: (n, l, label, x_col, y_pos, capacity)
subshells = [
    (1,0,'1s',1,0.5,2),
    (2,0,'2s',1,1.8,2),(2,1,'2p',2,1.8,6),
    (3,0,'3s',1,3.1,2),(3,1,'3p',2,3.1,6),(3,2,'3d',3,3.1,10),
    (4,0,'4s',1,4.4,2),(4,1,'4p',2,4.4,6),(4,2,'4d',3,4.4,10),(4,3,'4f',4,4.4,14),
    (5,0,'5s',1,5.7,2),(5,1,'5p',2,5.7,6),(5,2,'5d',3,5.7,10),
    (6,0,'6s',1,7.0,2),(6,1,'6p',2,7.0,6),
]

col_colors = {0:BLUE, 1:GREEN, 2:YELLOW, 3:RED}

for n,l,lbl,xcol,y,cap in subshells:
    xp = xcol * 1.6
    col = col_colors[l]
    ax.hlines(y, xp-0.5, xp+0.5, colors=col, linewidths=3.0)
    ax.text(xp, y+0.12, lbl, ha='center', va='bottom', color=col, fontsize=10)
    ax.text(xp, y-0.15, f'({cap})', ha='center', va='top', color=col, fontsize=8, alpha=0.7)

# Madelung diagonal arrows
# 1s, 2s, 2p, 3s, 3p, [4s, 3d], [4p, 5s, 4d], [4f,5p,6s,6p]...
# Draw diagonal sweep arrows
madelung_order = [
    (1,0),(2,0),(2,1),(3,0),(3,1),(4,0),(3,2),(4,1),(5,0),(4,2),
    (5,1),(6,0),(4,3),(5,2),(6,1)
]
x_map = {0:1.6,1:3.2,2:4.8,3:6.4}
y_map = {(n,l): y for n,l,lbl,xcol,y,cap in subshells}

for i in range(len(madelung_order)-1):
    n1,l1 = madelung_order[i]; n2,l2 = madelung_order[i+1]
    x1,y1 = x_map[l1], y_map[(n1,l1)]
    x2,y2 = x_map[l2], y_map[(n2,l2)]
    ax.annotate('', xy=(x2, y2+0.08), xytext=(x1, y1+0.08),
                arrowprops=dict(arrowstyle='->', color=ORANGE,
                                lw=0.9, mutation_scale=8, alpha=0.6,
                                connectionstyle='arc3,rad=-0.2'))

# Column labels
for l, lbl in [(0,'s'),(1,'p'),(2,'d'),(3,'f')]:
    ax.text(x_map[l], -0.3, f'$\\ell={l}$ ({lbl})',
            ha='center', color=col_colors[l], fontsize=11, fontweight='bold')

ax.text(0.0, 5.0, 'Energy\n$\\uparrow$', ha='center', color='#cdd6f4', fontsize=10,
        rotation=90)

# Legend for capacity
ax.text(7.5, 9.5, 'Subshell capacity:', color='#cdd6f4', fontsize=9)
for l,(lbl,cap) in enumerate([('s',2),('p',6),('d',10),('f',14)]):
    ax.text(7.5, 9.0-l*0.5, f'{lbl}: {cap} electrons',
            color=col_colors[l], fontsize=9)

ax.text(4.0, 10.0, 'Madelung Rule: fill in order of increasing $n+l$',
        ha='center', color=ORANGE, fontsize=10,
        bbox=dict(boxstyle='round,pad=0.3', fc=BG, ec=ORANGE, lw=0.8))

ax.set_title('Atomic Orbital Energies and Shell Filling', color='#cdd6f4', pad=8)
plt.tight_layout()
SAVE('fig02_shell_filling.png'); plt.show(); print('Fig 02 saved.')
```

## Fig 03 — Hund's Rules

```python

# Fig 03: Hund's rules applied to p^2 (carbon ground state)
# Show microstate table and selection of ground state

fig, ax = plt.subplots(figsize=(9, 5.5))
ax.set_xlim(0, 10); ax.set_ylim(0, 6)
ax.axis('off')

# ── Microstate table for p^2 ──────────────────────────────
# Show the three 2p orbitals (m_l = -1, 0, +1)
# and how to fill two electrons to maximise S then L

ax.text(5.0, 5.7, r'Hund\'s Rules: $p^2$ Configuration (Carbon $2p^2$)',
        ha='center', color='#cdd6f4', fontsize=12, fontweight='bold')

# Draw three orbital boxes
box_y = 4.0
ml_labels = ['-1', '0', '+1']
box_x = [1.5, 3.0, 4.5]
box_w, box_h = 1.1, 0.55

for x, lbl in zip(box_x, ml_labels):
    rect = plt.Rectangle((x-box_w/2, box_y-box_h/2), box_w, box_h,
                          fc='#2a2a3e', ec='#45475a', lw=1.5)
    ax.add_patch(rect)
    ax.text(x, box_y-box_h/2-0.18, f'$m_l={lbl}$',
            ha='center', color='#cdd6f4', fontsize=9)

# Fill: one up-spin in m_l=-1, one up-spin in m_l=0 (Hund rule 1: max S=1)
for x_e in [box_x[0], box_x[1]]:
    ax.annotate('', xy=(x_e, box_y+0.15), xytext=(x_e, box_y-0.15),
                arrowprops=dict(arrowstyle='->', color=BLUE, lw=2.0,
                                mutation_scale=10))

ax.text(3.0, box_y+0.55, r'Rule 1: max $S=1$ ($\uparrow\uparrow$, parallel spins)',
        ha='center', color=BLUE, fontsize=10,
        bbox=dict(boxstyle='round,pad=0.2', fc=BG, ec=BLUE, lw=0.7))

# ── Rules 2 & 3 results ────────────────────────────────────
ax.text(0.3, 2.8,
        'Rule 1: max $S$ → $S=1$ (triplet)\n'
        '    Both spins parallel ↑↑ in different orbitals',
        va='top', color=BLUE, fontsize=10)

ax.text(0.3, 2.0,
        'Rule 2: max $L$ → $L=1$\n'
        '    $m_{l_1}=-1,\\ m_{l_2}=0$: $M_L^{\\rm max}=1$ → $L=1$',
        va='top', color=GREEN, fontsize=10)

ax.text(0.3, 1.2,
        'Rule 3: less than half-filled → $J=|L-S|=|1-1|=0$',
        va='top', color=YELLOW, fontsize=10)

# Ground state box
ax.text(5.5, 2.2,
        r'Ground state: $^3P_0$',
        ha='center', va='center', color=RED, fontsize=16, fontweight='bold',
        bbox=dict(boxstyle='round,pad=0.5', fc='#2a2a3e', ec=RED, lw=2.0))

ax.text(5.5, 1.3,
        r'$^{2S+1}L_J = {}^3P_0$' '\n'
        r'$S=1,\ L=1,\ J=0$',
        ha='center', va='top', color=RED, fontsize=10)

# Arrow to ground state
ax.annotate('', xy=(5.5, 2.9), xytext=(5.5, 3.6),
            arrowprops=dict(arrowstyle='->', color=RED, lw=1.5, mutation_scale=12))
ax.text(5.5, 3.8, 'Apply all\nthree rules', ha='center', color='#aaa', fontsize=8)

plt.tight_layout()
SAVE('fig03_hunds_rules.png'); plt.show(); print('Fig 03 saved.')
```

## Fig 04 — Spectral Series

```python

# Fig 04: Hydrogen spectral series — Grotrian diagram + wavelengths

fig, axes = plt.subplots(1, 2, figsize=(11, 5.5))

# Left: Grotrian diagram
ax = axes[0]
ax.set_xlim(0, 6); ax.set_ylim(-0.5, 9)
ax.axis('off')

# Energy levels (relative, in eV, shifted so n=1 is at y=0)
# E_n = -13.6/n^2 eV
E_n = {n: -13.6/n**2 for n in range(1, 7)}
E_inf = 0.0   # ionisation
E_base = E_n[1]
y_scale = 8.0 / (E_inf - E_base)

def E_to_y(E):
    return (E - E_base) * y_scale

# Draw levels
x_left, x_right = 0.4, 4.0
for n in range(1, 7):
    y = E_to_y(E_n[n])
    lw = 2.5 if n <= 3 else 1.5
    ax.hlines(y, x_left, x_right, colors='#cdd6f4', linewidths=lw)
    ax.text(x_right+0.1, y, f'$n={n}$', va='center', color='#cdd6f4', fontsize=9)
    ax.text(x_left-0.08, y, f'{E_n[n]:.2f} eV', va='center', ha='right',
            color='#aaa', fontsize=7.5)

# Draw transitions
series = {
    'Lyman':  {'nf':1, 'ni_list':[2,3,4,5], 'color':PURPLE, 'x':1.0},
    'Balmer': {'nf':2, 'ni_list':[3,4,5,6], 'color':RED,    'x':2.0},
    'Paschen':{'nf':3, 'ni_list':[4,5,6],   'color':ORANGE, 'x':3.0},
}
for name, info in series.items():
    nf = info['nf']; col = info['color']; x_tr = info['x']
    for ni in info['ni_list']:
        y1 = E_to_y(E_n[nf]); y2 = E_to_y(E_n[ni])
        ax.annotate('', xy=(x_tr, y1+0.08), xytext=(x_tr, y2-0.08),
                    arrowprops=dict(arrowstyle='->', color=col, lw=1.4,
                                    mutation_scale=8))
    # Series label
    ax.text(x_tr, E_to_y(E_n[nf])-0.4, name, ha='center',
            color=col, fontsize=8.5, fontweight='bold')

ax.set_title('Grotrian Diagram', color='#cdd6f4', pad=8)

# Right: wavelength spectrum bars
ax2 = axes[1]
ax2.set_facecolor(BG)
ax2.set_xlim(80, 2000); ax2.set_ylim(0, 3.5)
ax2.set_xlabel(r'Wavelength (nm)', fontsize=11)
ax2.axis('off')

Ry_eV = 13.6
hc_eV_nm = 1240.0

series_data = [
    ('Lyman\n(UV)', 1, range(2,8), PURPLE, 0.6),
    ('Balmer\n(Vis/UV)', 2, range(3,9), RED, 1.5),
    ('Paschen\n(IR)', 3, range(4,9), ORANGE, 2.4),
]

for name, nf, ni_range, col, y_bar in series_data:
    waves = []
    for ni in ni_range:
        E_ph = Ry_eV*(1/nf**2 - 1/ni**2)
        if E_ph > 0: waves.append(hc_eV_nm / E_ph)
    for wl in waves:
        ax2.axvline(wl, y_bar-0.3, y_bar+0.3, color=col, lw=2.0, alpha=0.9)
    ax2.text(min(waves)-5, y_bar, name, ha='right', va='center',
             color=col, fontsize=9)

# Visible range shading
ax2.axvspan(380, 700, color='white', alpha=0.05)
ax2.text(540, 3.15, 'Visible', ha='center', color='#aaa', fontsize=8)

# Wavelength axis
ax2.axhline(0.1, color='#45475a', lw=1.5)
for wl in [100, 200, 400, 600, 800, 1000, 1500, 2000]:
    ax2.axvline(wl, 0.05, 0.12, color='#45475a', lw=0.8)
    ax2.text(wl, 0.0, str(wl), ha='center', va='top', color='#aaa', fontsize=7)

ax2.set_xlim(80, 2100)
ax2.set_title('Spectral Lines by Series', color='#cdd6f4', pad=8)

plt.suptitle('Hydrogen Spectral Series', color='#cdd6f4', fontsize=13, y=1.02)
plt.tight_layout()
SAVE('fig04_spectral_series.png'); plt.show(); print('Fig 04 saved.')
```

## Fig 05 — Zeeman Effect

```python

# Fig 05: Zeeman effect — normal and anomalous
# Left: level splitting vs B for a P -> S transition
# Right: sodium D lines splitting

fig, axes = plt.subplots(1, 2, figsize=(11, 5.5))

B = np.linspace(0, 1.0, 200)
mu_B = 1.0  # units

# ── Left: Normal Zeeman (S=0, L=1 -> L=0) ──────────────────
ax = axes[0]

# Upper level: L=1, J=1, g_J=1; m_J = -1, 0, +1
# Lower level: L=0, J=0, g_J = undefined (only m_J=0 exists)
E_upper = {1: mu_B*B, 0: 0*B, -1: -mu_B*B}
E_lower = {0: 0*B}

for mJ, col in [(1,RED),(0,'#cdd6f4'),(-1,BLUE)]:
    ax.plot(B, E_upper[mJ] + 3.0, color=col, lw=2.2)
    label = f'$m_J={mJ:+d}$' if mJ!=0 else '$m_J=0$'
    ax.text(B[-1]+0.02, E_upper[mJ][-1]+3.0, label, va='center',
            color=col, fontsize=8)

ax.plot(B, E_lower[0]+0.0, color=GREEN, lw=2.5)
ax.text(B[-1]+0.02, 0.0, '$m_J=0$', va='center', color=GREEN, fontsize=8)

# Selection rules: delta m_J = 0, +/-1
for dm, col_arr, ls in [(0,'#cdd6f4',':'),(+1,RED,'--'),(-1,BLUE,'--')]:
    ax.annotate('', xy=(0.5, dm*0.5+3.0), xytext=(0.5, 0.0),
                arrowprops=dict(arrowstyle='->', color=col_arr,
                                lw=1.2, linestyle=ls, mutation_scale=8))

ax.axvline(0, color='#45475a', lw=0.8, ls=':')
ax.set_xlabel(r'Magnetic field $B$', fontsize=11)
ax.set_ylabel(r'Energy', fontsize=11)
ax.set_title('Normal Zeeman Effect ($S=0$)\n$^1P_1 \to {}^1S_0$', color='#cdd6f4')
ax.set_ylim(-1, 5); ax.set_xlim(-0.05, 1.2)
ax.text(0.02, 1.5, 'Upper: $^1P_1$\n$g_J=1$', color='#cdd6f4', fontsize=8)
ax.text(0.02, -0.5, 'Lower: $^1S_0$', color=GREEN, fontsize=8)
ax.grid(True, alpha=0.15)

# ── Right: Anomalous Zeeman — Na D lines (2p -> 2s) ─────────
ax2 = axes[1]

# D2: 2P_{3/2} -> 2S_{1/2}  (J=3/2 -> J=1/2)
# D1: 2P_{1/2} -> 2S_{1/2}  (J=1/2 -> J=1/2)
# g_J for 2S_{1/2}: L=0,S=1/2,J=1/2 -> g=2
# g_J for 2P_{1/2}: L=1,S=1/2,J=1/2 -> g=2/3
# g_J for 2P_{3/2}: L=1,S=1/2,J=3/2 -> g=4/3

g_S12  = 2.0
g_P12  = 2.0/3
g_P32  = 4.0/3

# Base energies (schematic)
E_S12_base = 0.0
E_P12_base = 4.0
E_P32_base = 4.8   # D2 higher than D1

# mJ values and colours
cols_mj = {-1.5:PURPLE,-0.5:BLUE,0.5:GREEN,1.5:RED,-1:PURPLE,1:RED,0:'#cdd6f4'}

for mJ in [-0.5, 0.5]:
    ax2.plot(B, E_S12_base + g_S12*mJ*B, color=cols_mj[mJ], lw=2.)
    ax2.text(1.02, E_S12_base + g_S12*mJ*B[-1],
             f'$m_J={mJ:+.1f}$', va='center', color=cols_mj[mJ], fontsize=7.5)

for mJ in [-0.5, 0.5]:
    ax2.plot(B, E_P12_base + g_P12*mJ*B, color=cols_mj[mJ], lw=2., ls='--')

for mJ in [-1.5, -0.5, 0.5, 1.5]:
    ax2.plot(B, E_P32_base + g_P32*mJ*B, color=cols_mj[mJ], lw=2.)

ax2.axvline(0, color='#45475a', lw=0.8, ls=':')
ax2.text(0.02, 1.5, r'$2S_{1/2}$  ($g_J=2$)', color='#cdd6f4', fontsize=9)
ax2.text(0.02, 4.1, r'$2P_{1/2}$  ($g_J=2/3$)', color='#cdd6f4', fontsize=9)
ax2.text(0.02, 5.3, r'$2P_{3/2}$  ($g_J=4/3$)', color='#cdd6f4', fontsize=9)
ax2.annotate('D1', xy=(0.5, (E_P12_base+E_S12_base)/2),
             color=YELLOW, fontsize=11, fontweight='bold')
ax2.annotate('D2', xy=(0.7, (E_P32_base+E_S12_base)/2),
             color=ORANGE, fontsize=11, fontweight='bold')
ax2.set_xlabel(r'Magnetic field $B$', fontsize=11)
ax2.set_ylabel(r'Energy', fontsize=11)
ax2.set_title('Anomalous Zeeman: Na $D_1$, $D_2$ Lines', color='#cdd6f4')
ax2.set_ylim(-0.5, 7); ax2.set_xlim(-0.05, 1.2)
ax2.grid(True, alpha=0.15)

plt.suptitle('Zeeman Effect: Normal and Anomalous', color='#cdd6f4', fontsize=12, y=1.02)
plt.tight_layout()
SAVE('fig05_zeeman.png'); plt.show(); print('Fig 05 saved.')
```

## Fig 06 — Stark Effect

```python

# Fig 06: Stark effect — linear (n=2) and quadratic (n=1)

fig, axes = plt.subplots(1, 2, figsize=(11, 5))

E_field = np.linspace(0, 1.0, 300)  # in units of e*a0*E_field / E_Rydberg

# ── Left: Linear Stark effect for n=2 ───────────────────────
# Degenerate perturbation in {2s, 2p_0, 2p_+1, 2p_-1}
# Only 2s and 2p_0 mix (selection rules on m_l for z-field)
# Eigenvalues: E = E_2 +/- 3*e*a0*epsilon (and two unchanged 2p+-1)
# In units where e*a0 = 1

E2 = 0.0   # reference
slope = 3.0

ax = axes[0]
# The four n=2 levels
ax.plot(E_field, E2 + slope*E_field,    color=RED,    lw=2.5,
        label=r'$\Delta E = +3ea_0\mathcal{E}$')
ax.plot(E_field, E2 - slope*E_field,    color=BLUE,   lw=2.5,
        label=r'$\Delta E = -3ea_0\mathcal{E}$')
ax.plot(E_field, E2 + 0*E_field,        color=GREEN,  lw=2.0, ls='--',
        label=r'$\Delta E = 0$  ($m_l=\pm1$, degenerate)')
ax.plot(E_field, E2 + 0*E_field+0.02,   color=GREEN,  lw=2.0, ls='--')

ax.axvline(0, color='#45475a', lw=0.8, ls=':')
ax.axhline(0, color='#45475a', lw=0.6, ls=':')
ax.set_xlabel(r'Electric field $\mathcal{E}$ (arb. units)', fontsize=11)
ax.set_ylabel(r'Energy shift $\Delta E$', fontsize=11)
ax.set_title('Linear Stark Effect ($n=2$)', color='#cdd6f4')
ax.legend(fontsize=8.5, facecolor=BG, edgecolor='#45475a', labelcolor='#cdd6f4')
ax.grid(True, alpha=0.2)
ax.text(0.05, 0.92, r'$2s$ and $2p_0$ mix via $\langle 2s|z|2p_0\rangle \neq 0$',
        transform=ax.transAxes, color=YELLOW, fontsize=8,
        bbox=dict(boxstyle='round,pad=0.2', fc=BG, ec=YELLOW, lw=0.7))

# ── Right: Quadratic Stark effect for n=1 ───────────────────
# Delta E = - alpha_pol / 2 * E^2
# For hydrogen 1s: alpha_pol = 9/2 * a0^3 (in atomic units)
alpha_pol = 4.5   # a.u.

ax2 = axes[1]
E_shift = -0.5 * alpha_pol * E_field**2

ax2.plot(E_field, E_shift, color=BLUE, lw=2.5, label=r'$\Delta E = -\frac{1}{2}\alpha_{\rm pol}\mathcal{E}^2$')
ax2.axhline(0, color='#45475a', lw=0.8, ls=':')

# Induced dipole moment = alpha_pol * E
d_induced = alpha_pol * E_field
ax2_r = ax2.twinx()
ax2_r.plot(E_field, d_induced, color=YELLOW, lw=2.0, ls='--',
           label=r'$\langle d\rangle = \alpha_{\rm pol}\mathcal{E}$')
ax2_r.set_ylabel(r'Induced dipole $\langle d\rangle$ (a.u.)', color=YELLOW, fontsize=10)
ax2_r.tick_params(colors=YELLOW)
ax2_r.spines['right'].set_color(YELLOW)

ax2.set_xlabel(r'Electric field $\mathcal{E}$ (arb. units)', fontsize=11)
ax2.set_ylabel(r'Energy shift $\Delta E$ (a.u.)', fontsize=11)
ax2.set_title('Quadratic Stark Effect ($n=1$)', color='#cdd6f4')

lines1, labels1 = ax2.get_legend_handles_labels()
lines2, labels2 = ax2_r.get_legend_handles_labels()
ax2.legend(lines1+lines2, labels1+labels2, fontsize=9,
           facecolor=BG, edgecolor='#45475a', labelcolor='#cdd6f4')
ax2.grid(True, alpha=0.2)
ax2.text(0.25, 0.12, r'Polarisability $\alpha_{\rm pol}=\frac{9}{2}a_0^3$ (H atom)',
         transform=ax2.transAxes, color=BLUE, fontsize=8,
         bbox=dict(boxstyle='round,pad=0.2', fc=BG, ec=BLUE, lw=0.7))

plt.suptitle('Stark Effect: Linear ($n=2$) and Quadratic ($n=1$)',
             color='#cdd6f4', fontsize=12, y=1.02)
plt.tight_layout()
SAVE('fig06_stark.png'); plt.show(); print('Fig 06 saved.')
```

## Fig 07 — Hyperfine Structure

```python

# Fig 07: Hydrogen hyperfine structure — n=1 and n=2 levels
# 21 cm line between F=0 and F=1 of 1s

fig, ax = plt.subplots(figsize=(9, 6.5))
ax.set_xlim(0, 9); ax.set_ylim(-0.5, 8.5)
ax.axis('off')

def draw_hf_level(ax, x1, x2, y, color, lw=2., ls='-'):
    ax.hlines(y, x1, x2, colors=color, linewidths=lw, linestyles=ls)

def label_r(ax, x, y, text, color, fs=9):
    ax.text(x+0.08, y, text, va='center', color=color, fontsize=fs)

# ── 1s levels ─────────────────────────────────────────────
# J=1/2, I=1/2 -> F=0 (singlet) or F=1 (triplet)
y_1s_F1 = 1.8
y_1s_F0 = 0.5

draw_hf_level(ax, 0.5, 3.5, y_1s_F1, RED,  lw=2.8)
draw_hf_level(ax, 0.5, 3.5, y_1s_F0, BLUE, lw=2.8)

label_r(ax, 3.5, y_1s_F1, r'$1s_{1/2}$,  $F=1$  (triplet)', RED, 10)
label_r(ax, 3.5, y_1s_F0, r'$1s_{1/2}$,  $F=0$  (singlet)', BLUE, 10)

# 21 cm annotation
ax.annotate('', xy=(1.8, y_1s_F1-0.05), xytext=(1.8, y_1s_F0+0.05),
            arrowprops=dict(arrowstyle='<->', color=GREEN, lw=2.0))
ax.text(1.4, (y_1s_F0+y_1s_F1)/2,
        '21 cm\n1420 MHz\n$\\Delta E = 5.9\\times10^{-6}$ eV',
        ha='right', va='center', color=GREEN, fontsize=9,
        bbox=dict(boxstyle='round,pad=0.3', fc=BG, ec=GREEN, lw=0.8))

# ── 2s_{1/2} hyperfine ────────────────────────────────────
y_2s_F1 = 5.8
y_2s_F0 = 5.2

draw_hf_level(ax, 0.5, 3.5, y_2s_F1, RED,  lw=2.)
draw_hf_level(ax, 0.5, 3.5, y_2s_F0, BLUE, lw=2.)
label_r(ax, 3.5, y_2s_F1, r'$2s_{1/2}$,  $F=1$', RED, 9)
label_r(ax, 3.5, y_2s_F0, r'$2s_{1/2}$,  $F=0$', BLUE, 9)

# ── 2p_{1/2} hyperfine ────────────────────────────────────
y_2p12_F1 = 5.5
y_2p12_F0 = 4.9

draw_hf_level(ax, 0.5, 3.5, y_2p12_F1, YELLOW, lw=2.)
draw_hf_level(ax, 0.5, 3.5, y_2p12_F0, PURPLE, lw=2.)
label_r(ax, 3.5, y_2p12_F1, r'$2p_{1/2}$,  $F=1$', YELLOW, 9)
label_r(ax, 3.5, y_2p12_F0, r'$2p_{1/2}$,  $F=0$', PURPLE, 9)

# ── 2p_{3/2} hyperfine ────────────────────────────────────
for i, (F, col) in enumerate([(2,RED),(1,'#cdd6f4'),(0,BLUE),(-1,PURPLE)]):
    if F >= 0:
        yp = 7.5 - i*0.38
        draw_hf_level(ax, 0.5, 3.5, yp, col, lw=2.)
        label_r(ax, 3.5, yp, f'$2p_{{3/2}}$,  $F={F}$', col, 9)

# Lamb shift arrow between 2s_{1/2} and 2p_{1/2}
# ax.annotate('', xy=(0.3, y_2p12_F1), xytext=(0.3, y_2s_F0),
#             arrowprops=dict(arrowstyle='<->', color=ORANGE, lw=1.5))
ax.text(0.15, (y_2p12_F1+y_2s_F0)/2,
        'Lamb\n1057\nMHz', ha='center', va='center',
        color=ORANGE, fontsize=7.5)

# Scale labels
ax.text(0.1, 3.2, r'$\sim10^{10}$' '\nHz\n(fine)', ha='center',
        color='#aaa', fontsize=8, rotation=90)

ax.set_title('Hydrogen Hyperfine Structure\n'
             r'($n=1$: 21 cm line; $n=2$: Lamb shift + hyperfine)',
             color='#cdd6f4', pad=10)

# Radio astronomy note
ax.text(4.5, 0.5,
        '21 cm ($F=1\to F=0$):\nmost important radio astronomy line\n'
        'traces neutral H throughout the Galaxy',
        color=GREEN, fontsize=9,
        bbox=dict(boxstyle='round,pad=0.4', fc='#2a2a3e', ec=GREEN, lw=1.0))

plt.tight_layout()
SAVE('fig07_hyperfine.png'); plt.show(); print('Fig 07 saved.')
```
