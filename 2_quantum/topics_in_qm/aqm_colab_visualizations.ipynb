## Plot 1 · Ch 1 — Landau Levels & LLL Wavefunctions

**Left**: Energy ladder $E_n = \hbar\omega_B(n+\tfrac{1}{2})$ — Zeeman split (spin ↑↓).  
**Right**: Lowest Landau Level (LLL) probability densities $|\psi_m|^2 \propto r^{2|m|}e^{-r^2/2\ell_B^2}$ in symmetric gauge for $m=0,1,2,3$.

```
# ── Check output dir exists ─────────────────────────────
import os; os.makedirs(OUTDIR, exist_ok=True)
print(f"Saving to: {os.path.abspath(OUTDIR)}")
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import Circle, FancyArrowPatch
from mpl_toolkits.mplot3d import Axes3D
import os, zipfile

BG='#1e1e2e'; FG='#cdd6f4'; EDGE='#45475a'
BLUE='#7ec8e3'; YELLOW='#f9c74f'; GREEN='#90ee90'
RED='#e07070'; PURPLE='#c792ea'; ORANGE='#ffb347'

plt.rcParams.update({
    'figure.facecolor':BG,'axes.facecolor':BG,
    'text.color':FG,'axes.labelcolor':FG,
    'xtick.color':FG,'ytick.color':FG,
    'axes.edgecolor':EDGE,'grid.color':EDGE,'grid.alpha':0.28,
    'font.family':'DejaVu Sans','font.size':12,
    'axes.titlesize':13,'axes.titleweight':'bold',
    'axes.titlecolor':BLUE,'legend.framealpha':0.3,
    'legend.edgecolor':EDGE,
})

OUTDIR = 'aqm_figures'
os.makedirs(OUTDIR, exist_ok=True)

def save(name, fig=None):
    f = fig or plt.gcf()
    path = f'{OUTDIR}/{name}'
    f.savefig(path, dpi=180, bbox_inches='tight', facecolor=BG)
    print(f'  saved -> {path}')
    plt.show()


fig = plt.figure(figsize=(15, 5.5))
fig.patch.set_facecolor(BG)

# ── Left: Zeeman-split Landau levels ─────────────────────────────
ax1 = fig.add_subplot(1, 6, (1, 2))
ax1.set_facecolor(BG)
ax1.set_xlim(0, 6)
ax1.set_ylim(-0.5, 5.8)
ax1.axis('off')

x_dn, x_up = (0.3, 2.6), (3.4, 5.7)   # x spans for each spin column

for n in range(5):
    E_dn = float(n)          # ↓: n ℏω_B
    E_up = float(n + 1)      # ↑: (n+1) ℏω_B
    ax1.hlines(E_dn, *x_dn, colors=BLUE, lw=3.0, zorder=3)
    ax1.hlines(E_up, *x_up, colors=RED,  lw=3.0, zorder=3)
    ax1.text(x_dn[0]-0.12, E_dn, f'$n={n}$', va='center',
             ha='right', color=BLUE, fontsize=9)

# Degenerate pairs: dashed connectors
for n in range(1, 5):
    ax1.hlines(float(n), x_dn[1], x_up[0],
               colors=GREEN, lw=1.0, ls='--', alpha=0.55)

# Column headers
ax1.text(np.mean(x_dn), 5.5, 'Spin ↓', color=BLUE,
         ha='center', fontsize=11, fontweight='bold')
ax1.text(np.mean(x_up), 5.5, 'Spin ↑', color=RED,
         ha='center', fontsize=11, fontweight='bold')

# LLL annotation
ax1.annotate('LLL: only ↓\nat $E=0$',
             xy=(np.mean(x_dn), 0), xytext=(np.mean(x_dn), -0.42),
             ha='center', color=GREEN, fontsize=8.5,
             arrowprops=dict(arrowstyle='->', color=GREEN, lw=1.0))

# ℏω_B spacing arrow
ax1.annotate('', xy=(1.45, 2.0), xytext=(1.45, 1.0),
             arrowprops=dict(arrowstyle='<->', color=YELLOW, lw=1.5,
                             mutation_scale=12))
ax1.text(1.6, 1.5, r'$\hbar\omega_B$', color=YELLOW, fontsize=10, va='center')

# Energy axis
ax1.annotate('', xy=(0.15, 5.6), xytext=(0.15, -0.2),
             arrowprops=dict(arrowstyle='->', color=FG, lw=1.1))
ax1.text(0.05, 5.6, '$E$', color=FG, fontsize=13, va='top', style='italic')
ax1.set_title('Zeeman-Split Landau Levels  ($g_e=2$)',
              color=BLUE, fontsize=11, pad=6)

# ── Right: LLL wavefunctions (m=0,1,2,3) ─────────────────────────
L_box = 4.2
xy_arr = np.linspace(-L_box, L_box, 400)
X, Y = np.meshgrid(xy_arr, xy_arr)
r2 = X**2 + Y**2

for idx, m in enumerate([0, 1, 2, 3]):
    ax = fig.add_subplot(1, 6, 3 + idx)
    ax.set_facecolor(BG)
    psi_sq = r2**m * np.exp(-r2 / 2.0)   # ∝ r^{2m} e^{-r²/2ℓ_B²}, ℓ_B=1
    psi_sq /= psi_sq.max()
    ax.contourf(X, Y, psi_sq, levels=55, cmap='magma', vmin=0, vmax=1)
    ax.set_aspect('equal')
    ax.set_xlim(-L_box, L_box)
    ax.set_ylim(-L_box, L_box)
    ax.set_xticks([]); ax.set_yticks([])
    for sp in ax.spines.values(): sp.set_edgecolor(EDGE)
    if m > 0:
        r_peak = np.sqrt(m)   # peak radius in units of ℓ_B
        circ = Circle((0,0), r_peak, fill=False,
                      edgecolor=BLUE, lw=1.2, ls='--', alpha=0.8)
        ax.add_patch(circ)
        ax.text(r_peak*0.72, r_peak*0.72, r'$r_{\rm peak}$',
                color=BLUE, fontsize=7)
    ax.set_title(f'$m = {m}$', color=BLUE, fontsize=11, pad=3)

fig.suptitle(
    r'LLL wavefunctions $|\psi_m|^2 \propto r^{2|m|}e^{-r^2/2\ell_B^2}$'
    r'  (symmetric gauge)',
    color=FG, fontsize=11, y=1.01)
plt.tight_layout(pad=0.6)
save('plot1_landau_LLL.png', fig)
plt.close()
```

## Plot 2 · Ch 2 — Band Structure: NFE vs Tight-Binding

**Left**: Nearly-free electron model — free electron parabola folded into the first BZ; gaps open at zone boundaries.  
**Right**: Tight-binding dispersion $E(k)=E_0-2t\cos(ka)$ for several hopping values $t$; shows bandwidth $4t$ and effective mass at band bottom.

```
# ── Check output dir exists ─────────────────────────────
import os; os.makedirs(OUTDIR, exist_ok=True)
print(f"Saving to: {os.path.abspath(OUTDIR)}")
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import Circle, FancyArrowPatch
from mpl_toolkits.mplot3d import Axes3D
import os, zipfile

BG='#1e1e2e'; FG='#cdd6f4'; EDGE='#45475a'
BLUE='#7ec8e3'; YELLOW='#f9c74f'; GREEN='#90ee90'
RED='#e07070'; PURPLE='#c792ea'; ORANGE='#ffb347'

plt.rcParams.update({
    'figure.facecolor':BG,'axes.facecolor':BG,
    'text.color':FG,'axes.labelcolor':FG,
    'xtick.color':FG,'ytick.color':FG,
    'axes.edgecolor':EDGE,'grid.color':EDGE,'grid.alpha':0.28,
    'font.family':'DejaVu Sans','font.size':12,
    'axes.titlesize':13,'axes.titleweight':'bold',
    'axes.titlecolor':BLUE,'legend.framealpha':0.3,
    'legend.edgecolor':EDGE,
})

OUTDIR = 'aqm_figures'
os.makedirs(OUTDIR, exist_ok=True)

def save(name, fig=None):
    f = fig or plt.gcf()
    path = f'{OUTDIR}/{name}'
    f.savefig(path, dpi=180, bbox_inches='tight', facecolor=BG)
    print(f'  saved -> {path}')
    plt.show()


fig, axes = plt.subplots(1, 3, figsize=(13, 5))
fig.patch.set_facecolor(BG)

# ── Panel 1: Nearly-free electrons (extended zone + folded) ──────
ax = axes[0]
k_full = np.linspace(-3*np.pi, 3*np.pi, 2000)  # in units of 1/a
E_free = k_full**2 / 2.0   # free electron E = ℏ²k²/2m, set ℏ²/2m=1, a=1

ax.plot(k_full/np.pi, E_free, color=EDGE, lw=1.2, ls='--',
        label='Free electron', alpha=0.7)

# Folded bands (reduced zone): fold back at G = 2πn/a
colors_band = [BLUE, YELLOW, GREEN, RED, PURPLE]
for n in range(-2, 3):
    k_rz = np.linspace(-np.pi, np.pi, 600)
    k_ext = k_rz + 2*np.pi*n
    E_band = k_ext**2 / 2.0
    ax.plot(k_rz/np.pi, E_band, color=colors_band[n+2], lw=2.0, alpha=0.85)

# BZ boundaries
for xb in [-1, 1]:
    ax.axvline(xb, color=EDGE, lw=0.8, ls=':')
ax.axhline(0, color=EDGE, lw=0.5)

# Band gap annotation at zone boundary
E_gap_lo, E_gap_hi = np.pi**2/2 - 1.0, np.pi**2/2 + 1.0  # schematic gap
ax.annotate('', xy=(1.05, E_gap_hi), xytext=(1.05, E_gap_lo),
            arrowprops=dict(arrowstyle='<->', color=RED, lw=1.5,
                            mutation_scale=10))
ax.text(1.15, (E_gap_lo+E_gap_hi)/2, r'gap $\!=\!2|V_G|$',
        color=RED, fontsize=8, va='center')

ax.set_xlabel(r'$ka/\pi$', fontsize=11)
ax.set_ylabel(r'Energy $E$ (arb.)', fontsize=11)
ax.set_title('Nearly-Free Electrons\n(Reduced Zone Scheme)', color=BLUE)
ax.set_xlim(-1.35, 1.35)
ax.set_ylim(-0.3, 28)
ax.set_xticks([-1, 0, 1], [r'$-\pi/a$', '0', r'$\pi/a$'])
ax.legend(fontsize=9, labelcolor=FG)
ax.grid(True)

# ── Panel 2: Tight-binding dispersion ────────────────────────────
ax2 = axes[1]
k = np.linspace(-np.pi, np.pi, 500)
t_vals   = [0.5, 1.0, 2.0, 3.5]
t_colors = [PURPLE, BLUE, YELLOW, GREEN]

for t, col in zip(t_vals, t_colors):
    E = -2*t*np.cos(k)
    ax2.plot(k/np.pi, E, color=col, lw=2.2, label=f'$t={t}$')

ax2.axhline(0, color=EDGE, lw=0.5)
for xb in [-1, 1]:
    ax2.axvline(xb, color=EDGE, lw=0.8, ls=':')

# Bandwidth annotation for t=2
ax2.annotate('', xy=(0.0, -4.0), xytext=(0.0, 4.0),
             arrowprops=dict(arrowstyle='<->', color=YELLOW,
                             lw=1.5, mutation_scale=10))
ax2.text(0.12, 0, 'BW$=4t$', color=YELLOW, fontsize=9, va='center')

ax2.set_xlabel(r'$ka/\pi$', fontsize=11)
ax2.set_ylabel(r'$E(k)-E_0$  (units of $t$)', fontsize=11)
ax2.set_title('Tight-Binding Dispersion\n$E(k)=E_0-2t\\cos(ka)$', color=BLUE)
ax2.set_xlim(-1.1, 1.1)
ax2.set_xticks([-1, 0, 1], [r'$-\pi/a$', '0', r'$\pi/a$'])
ax2.legend(fontsize=9, labelcolor=FG, loc='upper center')
ax2.grid(True)

# ── Panel 3: DOS schematic — NFE vs TB ───────────────────────────
ax3 = axes[2]
E_dos = np.linspace(-4.5, 4.5, 800)

# TB DOS for 1D: g(E) ∝ 1/√(1-(E/2t)²) — van Hove singularities
t0 = 2.0
mask = np.abs(E_dos) < 2*t0 - 0.01
DOS_tb = np.zeros_like(E_dos)
DOS_tb[mask] = 1.0 / np.sqrt(1 - (E_dos[mask]/(2*t0))**2 + 1e-6)
DOS_tb = np.clip(DOS_tb, 0, 6)
DOS_tb /= DOS_tb.max()

# Free electron DOS for 1D: g(E) ∝ 1/√E
mask2 = E_dos > 0.05
DOS_free = np.zeros_like(E_dos)
DOS_free[mask2] = 1.0 / np.sqrt(E_dos[mask2])
DOS_free /= DOS_free.max()

ax3.plot(DOS_tb,   E_dos, color=BLUE,   lw=2.2, label='Tight-binding (1D)')
ax3.plot(DOS_free, E_dos+0.5, color=YELLOW, lw=2.0,
         ls='--', label='Free electron (1D)', alpha=0.8)

ax3.axhline(0, color=EDGE, lw=0.5)
ax3.fill_betweenx(E_dos, 0, DOS_tb,
                  where=np.abs(E_dos)<2*t0-0.01,
                  color=BLUE, alpha=0.10)

# Van Hove singularities
for y_vH, label in [(-2*t0+0.05, 'van Hove\nsingularity'), (2*t0-0.05, '')]:
    ax3.axhline(y_vH, color=RED, lw=0.8, ls='--', alpha=0.6)
ax3.text(0.55, -2*t0+0.3, 'van Hove\nsingularity', color=RED, fontsize=8)

ax3.set_xlabel('DOS $g(E)$ (arb.)', fontsize=11)
ax3.set_ylabel('Energy $E$ (arb.)', fontsize=11)
ax3.set_title('Density of States (1D)', color=BLUE)
ax3.set_xlim(-0.05, 1.15)
ax3.set_ylim(-4.8, 4.8)
ax3.legend(fontsize=9, labelcolor=FG)
ax3.grid(True)

plt.tight_layout(pad=1.2)
save('plot2_band_structure.png', fig)
plt.close()
```

## Plot 3 · Ch 3 — Electron Dynamics: Semi-classical & Bloch Oscillations

**Left**: Semi-classical velocity $v(k) = (1/\hbar)\,dE/dk$ and effective mass $m^*(k) = \hbar^2/(d^2E/dk^2)$ for a tight-binding band.  
**Centre**: Bloch oscillations — $k(t)$, $v(t)$, $x(t)$ in a DC electric field.  
**Right**: 2D Fermi surface for a square tight-binding lattice at various fillings.

```
# ── Check output dir exists ─────────────────────────────
import os; os.makedirs(OUTDIR, exist_ok=True)
print(f"Saving to: {os.path.abspath(OUTDIR)}")
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import Circle, FancyArrowPatch
from mpl_toolkits.mplot3d import Axes3D
import os, zipfile

BG='#1e1e2e'; FG='#cdd6f4'; EDGE='#45475a'
BLUE='#7ec8e3'; YELLOW='#f9c74f'; GREEN='#90ee90'
RED='#e07070'; PURPLE='#c792ea'; ORANGE='#ffb347'

plt.rcParams.update({
    'figure.facecolor':BG,'axes.facecolor':BG,
    'text.color':FG,'axes.labelcolor':FG,
    'xtick.color':FG,'ytick.color':FG,
    'axes.edgecolor':EDGE,'grid.color':EDGE,'grid.alpha':0.28,
    'font.family':'DejaVu Sans','font.size':12,
    'axes.titlesize':13,'axes.titleweight':'bold',
    'axes.titlecolor':BLUE,'legend.framealpha':0.3,
    'legend.edgecolor':EDGE,
})

OUTDIR = 'aqm_figures'
os.makedirs(OUTDIR, exist_ok=True)

def save(name, fig=None):
    f = fig or plt.gcf()
    path = f'{OUTDIR}/{name}'
    f.savefig(path, dpi=180, bbox_inches='tight', facecolor=BG)
    print(f'  saved -> {path}')
    plt.show()


fig, axes = plt.subplots(1, 3, figsize=(13, 5))
fig.patch.set_facecolor(BG)

# ── Panel 1: velocity & effective mass ───────────────────────────
ax = axes[0]
k   = np.linspace(-np.pi, np.pi, 600)
t_tb = 1.0
E   = -2*t_tb*np.cos(k)
v   = 2*t_tb*np.sin(k)           # dE/dk  (ℏ=a=1)
m_inv = -2*t_tb*np.cos(k)        # d²E/dk²
# Effective mass = 1/m_inv (clip for display)
m_eff = np.where(np.abs(m_inv) > 0.15, 1.0/m_inv, np.nan)
m_eff_clipped = np.clip(m_eff, -8, 8)

ax2_twin = ax.twinx()
ax2_twin.set_facecolor(BG)
ax2_twin.tick_params(colors=FG)
ax2_twin.spines['right'].set_color(RED)
ax2_twin.yaxis.label.set_color(RED)

ax.plot(k/np.pi,  E,              color=BLUE,   lw=2.2, label='$E(k)$')
ax.plot(k/np.pi,  v,              color=YELLOW,  lw=2.0, label='$v(k)=dE/dk$', ls='--')
ax2_twin.plot(k/np.pi, m_eff_clipped, color=RED, lw=1.5,
              label=r'$m^*(k)$', ls=':', alpha=0.9)

ax.axhline(0, color=EDGE, lw=0.5)
ax.axvline(0, color=EDGE, lw=0.5, ls='--')
for xb in [-1, 1]: ax.axvline(xb, color=EDGE, lw=0.7, ls=':')

ax.set_xlabel(r'$ka/\pi$', fontsize=11)
ax.set_ylabel('$E$, $v$  (arb.)', fontsize=11)
ax2_twin.set_ylabel(r'$m^*(k)/m^*_0$', fontsize=10, color=RED)
ax.set_title('Band Velocity & Effective Mass', color=BLUE)
ax.set_xlim(-1.1, 1.1)
ax.set_xticks([-1,0,1],[r'$-\pi/a$','0',r'$\pi/a$'])
lines1, labs1 = ax.get_legend_handles_labels()
lines2, labs2 = ax2_twin.get_legend_handles_labels()
ax.legend(lines1+lines2, labs1+labs2, fontsize=9, labelcolor=FG, loc='upper left')
ax.grid(True)

# ── Panel 2: Bloch oscillations ───────────────────────────────────
ax2 = axes[1]
t_time  = np.linspace(0, 4*np.pi, 1000)   # time in units of ℏ/eEa
omega_B = 1.0                               # Bloch frequency = eEa/ℏ
k_t     = -omega_B * t_time                 # k(t) = k₀ - eEt/ℏ
v_t     = 2*t_tb * np.sin(k_t)             # v(t)
x_t     = (2*t_tb/omega_B) * np.cos(k_t)  # x(t) = -(2t/eE)cos(ωt)

ax2.plot(t_time/np.pi, k_t/np.pi,  color=PURPLE, lw=2.0, label=r'$k(t)\,/\,\pi$')
ax2.plot(t_time/np.pi, v_t/(2*t_tb), color=YELLOW, lw=2.0, label=r'$v(t)\,/\,v_{\max}$')
ax2.plot(t_time/np.pi, x_t/(2*t_tb/omega_B), color=GREEN, lw=2.0,
         label=r'$x(t)\,/\,x_{\max}$')

# π-period lines
for xp in [1, 2, 3, 4]:
    ax2.axvline(xp, color=EDGE, lw=0.7, ls='--')
ax2.axhline(0, color=EDGE, lw=0.5)

T_B = 2*np.pi/omega_B
ax2.annotate('', xy=(2.05, 1.15), xytext=(0.05, 1.15),
             arrowprops=dict(arrowstyle='<->', color=RED, lw=1.3, mutation_scale=10))
ax2.text(1.0, 1.22, r'$T_B=h/eEa$', color=RED, ha='center', fontsize=9)

ax2.set_xlabel(r'$t\,/\,(\pi/\omega_B)$', fontsize=11)
ax2.set_ylabel('Normalised units', fontsize=11)
ax2.set_title('Bloch Oscillations\n(DC electric field)', color=BLUE)
ax2.set_xlim(0, 4)
ax2.set_ylim(-1.4, 1.45)
ax2.set_xticks([0,1,2,3,4])
ax2.legend(fontsize=9, labelcolor=FG, loc='lower right')
ax2.grid(True)

# ── Panel 3: 2D Fermi surfaces (square lattice) ──────────────────
ax3 = axes[2]
kx = np.linspace(-np.pi, np.pi, 400)
ky = np.linspace(-np.pi, np.pi, 400)
KX, KY = np.meshgrid(kx, ky)
# 2D TB dispersion: E(kx,ky) = -2t(cos kx + cos ky)
E_2d = -2*t_tb*(np.cos(KX) + np.cos(KY))

# Plot Fermi surfaces at different fillings
E_F_vals  = [-3.0, -1.5,  0.0,  2.5]
FS_colors = [BLUE, YELLOW, GREEN, RED]
labels    = ['Quarter fill', 'Half fill (≈)',
             'Half fill ($E_F=0$)', 'Near-full fill']

for E_F, col, lab in zip(E_F_vals, FS_colors, labels):
    ax3.contour(KX/np.pi, KY/np.pi, E_2d, levels=[E_F],
                colors=[col], linewidths=2.0)
    # Dummy for legend
    ax3.plot([], [], color=col, lw=2, label=lab)

# BZ boundary (square)
sq_x = [-1, 1, 1, -1, -1]
sq_y = [-1,-1, 1,  1, -1]
ax3.plot(sq_x, sq_y, color=EDGE, lw=1.5, ls='--', alpha=0.7)

ax3.set_aspect('equal')
ax3.set_xlabel(r'$k_x a/\pi$', fontsize=11)
ax3.set_ylabel(r'$k_y a/\pi$', fontsize=11)
ax3.set_title('2D Fermi Surfaces\n(Square TB lattice)', color=BLUE)
ax3.legend(fontsize=8, labelcolor=FG, loc='upper right')
ax3.set_xlim(-1.25, 1.25); ax3.set_ylim(-1.25, 1.25)
ax3.grid(True)

plt.tight_layout(pad=1.2)
save('plot3_electron_dynamics.png', fig)
plt.close()
```

## Plot 4 · Ch 4 — Phonon Dispersion: Monatomic & Diatomic Chain

**Left**: 1D monatomic chain $\omega(k)=2\sqrt{\kappa/m}|\sin(ka/2)|$ — single acoustic branch.  
**Right**: 1D diatomic chain with mass ratio $M/m=3$ — acoustic + optical branches with a phonon gap at the zone boundary.  
**Inset**: Phonon occupation $\langle n\rangle = 1/(e^{\hbar\omega/k_BT}-1)$ vs $\hbar\omega/k_BT$.

```
# ── Check output dir exists ─────────────────────────────
import os; os.makedirs(OUTDIR, exist_ok=True)
print(f"Saving to: {os.path.abspath(OUTDIR)}")
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import Circle, FancyArrowPatch
from mpl_toolkits.mplot3d import Axes3D
import os, zipfile

BG='#1e1e2e'; FG='#cdd6f4'; EDGE='#45475a'
BLUE='#7ec8e3'; YELLOW='#f9c74f'; GREEN='#90ee90'
RED='#e07070'; PURPLE='#c792ea'; ORANGE='#ffb347'

plt.rcParams.update({
    'figure.facecolor':BG,'axes.facecolor':BG,
    'text.color':FG,'axes.labelcolor':FG,
    'xtick.color':FG,'ytick.color':FG,
    'axes.edgecolor':EDGE,'grid.color':EDGE,'grid.alpha':0.28,
    'font.family':'DejaVu Sans','font.size':12,
    'axes.titlesize':13,'axes.titleweight':'bold',
    'axes.titlecolor':BLUE,'legend.framealpha':0.3,
    'legend.edgecolor':EDGE,
})

OUTDIR = 'aqm_figures'
os.makedirs(OUTDIR, exist_ok=True)

def save(name, fig=None):
    f = fig or plt.gcf()
    path = f'{OUTDIR}/{name}'
    f.savefig(path, dpi=180, bbox_inches='tight', facecolor=BG)
    print(f'  saved -> {path}')
    plt.show()


fig, axes = plt.subplots(1, 3, figsize=(13, 5))
fig.patch.set_facecolor(BG)

kappa, m_mono = 1.0, 1.0
k_bz = np.linspace(-np.pi, np.pi, 600)

# ── Panel 1: Monatomic chain ─────────────────────────────────────
ax = axes[0]
omega_mono = 2*np.sqrt(kappa/m_mono)*np.abs(np.sin(k_bz/2))
omega_max  = 2*np.sqrt(kappa/m_mono)

ax.plot(k_bz/np.pi, omega_mono/omega_max,
        color=BLUE, lw=2.5, label='Acoustic')
ax.fill_between(k_bz/np.pi, 0, omega_mono/omega_max,
                color=BLUE, alpha=0.10)

# Sound velocity slope at k=0
v_s = np.sqrt(kappa/m_mono)   # dω/dk|_{k=0}
k_lin = np.linspace(0, np.pi/2, 100)
ax.plot(k_lin/np.pi, v_s*k_lin/omega_max,
        color=YELLOW, lw=1.5, ls='--', alpha=0.8, label=r'$\omega=v_s k$ (sound)')

# Zone boundary: group velocity = 0
ax.plot([1], [1], 'o', color=RED, ms=8, zorder=5)
ax.text(0.88, 1.04, r'$v_g=0$', color=RED, fontsize=9)

for xb in [-1,1]: ax.axvline(xb, color=EDGE, lw=0.8, ls=':')
ax.set_xlabel(r'$ka/\pi$', fontsize=11)
ax.set_ylabel(r'$\omega/\omega_{\max}$', fontsize=11)
ax.set_title('Monatomic Chain\n(Acoustic branch)', color=BLUE)
ax.legend(fontsize=9, labelcolor=FG)
ax.set_xlim(-1.15, 1.15); ax.set_ylim(-0.03, 1.18)
ax.set_xticks([-1,0,1],[r'$-\pi/a$','0',r'$\pi/a$'])
ax.grid(True)

# ── Panel 2: Diatomic chain ──────────────────────────────────────
ax2 = axes[1]
M_ratio = 3.0   # M/m
m_light = 1.0; M_heavy = M_ratio * m_light

# Full diatomic dispersion (unit cell = 2 atoms, BZ boundary at π/2a)
k_di = np.linspace(-np.pi/2, np.pi/2, 600)  # half-BZ (2a unit cell)
reduced = (np.sin(k_di))**2 / (M_heavy * m_light)
prefactor = kappa * (1/M_heavy + 1/m_light)
root = np.sqrt(prefactor**2 - 4*kappa**2 * reduced)
omega_ac  = np.sqrt((prefactor - root)/2)
omega_opt = np.sqrt((prefactor + root)/2)

omega_scale = np.sqrt(2*kappa/m_light)  # normalise

ax2.plot(k_di/np.pi*2, omega_ac /omega_scale, color=BLUE,   lw=2.5, label='Acoustic')
ax2.plot(k_di/np.pi*2, omega_opt/omega_scale, color=ORANGE, lw=2.5, label='Optical')
ax2.fill_between(k_di/np.pi*2, omega_ac/omega_scale, omega_opt/omega_scale,
                 color=RED, alpha=0.08, label='Phonon gap')

# Gap annotation at zone boundary
gap_lo = omega_ac[-1]/omega_scale
gap_hi = omega_opt[-1]/omega_scale
ax2.annotate('', xy=(1.1, gap_hi), xytext=(1.1, gap_lo),
             arrowprops=dict(arrowstyle='<->', color=RED, lw=1.5, mutation_scale=10))
ax2.text(1.18, (gap_lo+gap_hi)/2, 'gap', color=RED, va='center', fontsize=9)

for xb in [-1,1]: ax2.axvline(xb, color=EDGE, lw=0.8, ls=':')
ax2.set_xlabel(r'$ka/(\pi/2)$  (reduced BZ)', fontsize=10)
ax2.set_ylabel(r'$\omega/\omega_0$', fontsize=11)
ax2.set_title(f'Diatomic Chain  ($M/m={M_ratio:.0f}$)\nAcoustic + Optical branches',
              color=BLUE)
ax2.legend(fontsize=9, labelcolor=FG)
ax2.set_xlim(-1.25, 1.35); ax2.set_ylim(-0.03, 1.35)
ax2.set_xticks([-1,0,1])
ax2.grid(True)

# ── Panel 3: Bose–Einstein occupation + Debye heat capacity ──────
ax3 = axes[2]
x = np.linspace(0.05, 8, 500)   # x = ℏω/k_BT
n_BE = 1.0 / (np.exp(x) - 1)   # Planck / BE distribution

ax3.semilogy(x, n_BE, color=BLUE, lw=2.5, label=r'$\langle n\rangle = 1/(e^{\hbar\omega/k_BT}-1)$')
ax3.semilogy(x, np.exp(-x), color=YELLOW, lw=1.8, ls='--',
             label=r'Classical: $e^{-\hbar\omega/k_BT}$')
ax3.fill_between(x, n_BE, 1e-3, where=n_BE>1e-3, color=BLUE, alpha=0.08)

ax3.axvline(1, color=RED, lw=1.0, ls='--', alpha=0.7)
ax3.text(1.1, 5, r'$\hbar\omega=k_BT$', color=RED, fontsize=9)

# Debye T³ inset text
ax3.text(0.15, 0.012,
         'Debye model: $C_V$',
         color=GREEN, fontsize=8.5,
         bbox=dict(boxstyle='round,pad=0.4', fc=BG, ec=GREEN, lw=0.8))

ax3.set_xlabel(r'$\hbar\omega / k_BT$', fontsize=11)
ax3.set_ylabel(r'Phonon occupation $\langle n \rangle$', fontsize=11)
ax3.set_title('Bose–Einstein Distribution\n(Phonon Statistics)', color=BLUE)
ax3.legend(fontsize=9, labelcolor=FG)
ax3.set_xlim(0, 8); ax3.set_ylim(1e-3, 30)
ax3.grid(True)

plt.tight_layout(pad=1.2)
save('plot4_phonons.png', fig)
plt.close()
```

## Plot 5 · Ch 8 — Rabi Oscillations & Bloch Sphere

**Left**: Excited-state population $P_2(t)$ for resonant and off-resonant driving.  
**Centre**: Generalised Rabi frequency $\tilde\Omega = \sqrt{\Omega_R^2+\delta^2}$ vs detuning.

```
# ── Check output dir exists ─────────────────────────────
import os; os.makedirs(OUTDIR, exist_ok=True)
print(f"Saving to: {os.path.abspath(OUTDIR)}")
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D, proj3d
import matplotlib.lines as mlines
from PIL import Image
import os

BG='#1e1e2e'; FG='#cdd6f4'; EDGE='#45475a'
BLUE='#7ec8e3'; YELLOW='#f9c74f'; GREEN='#90ee90'
RED='#e07070'; PURPLE='#c792ea'
BG_RGB = np.array([30, 30, 46])

plt.rcParams.update({
    'figure.facecolor':BG,'axes.facecolor':BG,
    'text.color':FG,'axes.labelcolor':FG,
    'xtick.color':FG,'ytick.color':FG,
    'axes.edgecolor':EDGE,'grid.color':EDGE,'grid.alpha':0.28,
    'font.family':'DejaVu Sans','font.size':12,
    'axes.titlesize':13,'axes.titleweight':'bold',
    'axes.titlecolor':BLUE,'legend.framealpha':0.3,
    'legend.edgecolor':EDGE,
})

OUTDIR = 'aqm_figures'
os.makedirs(OUTDIR, exist_ok=True)

def save(name, fig=None):
    f = fig or plt.gcf()
    path = f'{OUTDIR}/{name}'
    f.savefig(path, dpi=180, bbox_inches='tight', facecolor=BG)
    print(f'  saved -> {path}')
    plt.show()

# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
# Plot 5a: Rabi oscillations (left 2 panels)
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
fig, axes = plt.subplots(1, 2, figsize=(10, 5))
fig.patch.set_facecolor(BG)

# Panel 1: Rabi oscillations
ax1 = axes[0]
t = np.linspace(0, 4*np.pi, 800)
Omega_R = 1.0
cases = [
    (0.0,       BLUE,   r'$\delta=0$ (resonant)'),
    (Omega_R,   YELLOW, r'$\delta=\Omega_R$'),
    (2*Omega_R, RED,    r'$\delta=2\Omega_R$'),
]
for delta, col, lab in cases:
    Ot = np.sqrt(Omega_R**2 + delta**2)
    P2 = (Omega_R/Ot)**2 * np.sin(Ot*t/2)**2
    ax1.plot(t, P2, color=col, lw=2.2, label=lab)

ax1.axvline(np.pi,   color=GREEN, lw=1.0, ls='--', alpha=0.6)
ax1.axvline(2*np.pi, color=GREEN, lw=1.0, ls='--', alpha=0.4)
ax1.text(np.pi,   1.05, r'$\pi$',   color=GREEN, ha='center', fontsize=10)
ax1.text(2*np.pi, 1.05, r'$2\pi$',  color=GREEN, ha='center', fontsize=10)
ax1.set_xlabel(r'Time $\Omega_R t$', fontsize=11)
ax1.set_ylabel(r'$P_2(t)$', fontsize=11)
ax1.set_title('Rabi Oscillations', color=BLUE)
ax1.set_xlim(0, 4*np.pi)
ax1.set_ylim(-0.04, 1.15)
ax1.set_xticks([0,np.pi,2*np.pi,3*np.pi,4*np.pi],
               ['0',r'$\pi$',r'$2\pi$',r'$3\pi$',r'$4\pi$'])
ax1.legend(fontsize=9, labelcolor=FG)
ax1.grid(True)

# Panel 2: generalised Rabi freq vs detuning
ax2 = axes[1]
delta_arr = np.linspace(-4, 4, 400)
Omega_tilde = np.sqrt(1 + delta_arr**2)
P2_max = 1.0 / Omega_tilde**2

ax2.plot(delta_arr, Omega_tilde, color=BLUE,   lw=2.2, label=r'$\tilde\Omega/\Omega_R$')
ax2.plot(delta_arr, P2_max,      color=YELLOW,  lw=2.0, ls='--',
         label=r'Max $P_2=(\Omega_R/\tilde\Omega)^2$')
ax2.fill_between(delta_arr, 0, P2_max, color=YELLOW, alpha=0.08)
ax2.axvline(0, color=GREEN, lw=0.8, ls='--')
ax2.text(0.15, 0.95, 'Resonance', color=GREEN, fontsize=9,
         transform=ax2.get_xaxis_transform())
ax2.set_xlabel(r'Detuning $\delta/\Omega_R$', fontsize=11)
ax2.set_ylabel('(arb.)', fontsize=11)
ax2.set_title(r'Generalised Rabi Freq. $\tilde\Omega=\sqrt{\Omega_R^2+\delta^2}$',
              color=BLUE, fontsize=11)
ax2.legend(fontsize=9, labelcolor=FG)
ax2.set_xlim(-4, 4); ax2.set_ylim(-0.05, 4.5)
ax2.grid(True)

plt.tight_layout(pad=1.2)
save('plot5a_rabi.png', fig)

# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
# Plot 5b: Bloch sphere (separate figure, canvas-buffer crop)
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
fig2 = plt.figure(figsize=(8, 7), dpi=150)
ax3 = fig2.add_axes([0.12, 0.04, 0.76, 0.92], projection='3d')
ax3.set_facecolor(BG); fig2.patch.set_facecolor(BG)

u = np.linspace(0, 2*np.pi, 30); v = np.linspace(0, np.pi, 16)
ax3.plot_wireframe(np.outer(np.cos(u),np.sin(v)),
                   np.outer(np.sin(u),np.sin(v)),
                   np.outer(np.ones(30),np.cos(v)),
                   color='#2d2d4e', lw=0.4, alpha=0.5)
eq = np.linspace(0, 2*np.pi, 150)
ax3.plot(np.cos(eq), np.sin(eq), 0, color=EDGE, lw=0.8, alpha=0.5)
ax3.quiver(0,0,-1.3, 0,0,2.6, color='#555', lw=1.0, arrow_length_ratio=0.05)

theta_p, phi_p = np.radians(50), np.radians(-30)
r_pure  = np.array([np.sin(theta_p)*np.cos(phi_p),
                    np.sin(theta_p)*np.sin(phi_p),
                    np.cos(theta_p)])
r_mixed = r_pure * 0.50

ax3.quiver(0,0,0,*r_pure,  color=BLUE,   lw=2.8, arrow_length_ratio=0.09)
ax3.quiver(0,0,0,*r_mixed, color=YELLOW, lw=2.8, arrow_length_ratio=0.14)
ax3.scatter([0],[0],[0], color=RED,    s=60, zorder=6)
ax3.scatter(*r_pure,    color=BLUE,   s=35, zorder=7)
ax3.scatter(*r_mixed,   color=YELLOW, s=35, zorder=7)

t_dec = np.linspace(0, 3*np.pi, 300)
decay = np.exp(-t_dec/(3*np.pi))
dec_x = decay*np.sin(theta_p)*np.cos(phi_p+t_dec)
dec_y = decay*np.sin(theta_p)*np.sin(phi_p+t_dec)
dec_z = np.cos(theta_p)*np.ones_like(t_dec)
ax3.plot(dec_x, dec_y, dec_z, color=GREEN, lw=2.0, alpha=0.9)
spiral_pt = np.array([dec_x[0], dec_y[0], dec_z[0]])

ax3.text( 0.08,  0.08,  1.18, r'$|\uparrow\rangle$',  color=FG, fontsize=11)
ax3.text(-0.08,  0.08, -1.28, r'$|\downarrow\rangle$', color=FG, fontsize=11)

ax3.set_xlim(-1.4,1.4); ax3.set_ylim(-1.4,1.4); ax3.set_zlim(-1.5,1.5)
ax3.set_axis_off()
ax3.view_init(elev=20, azim=-60)
ax3.set_title('Bloch Sphere: Pure, Mixed, Decoherence', color=BLUE, fontsize=12, pad=8)

# 1st draw: fix 3D coords
fig2.canvas.draw()

def to_fig(ax, fig, pt3d):
    x2, y2, _ = proj3d.proj_transform(*pt3d, ax.get_proj())
    xd, yd = ax.transData.transform((x2, y2))
    return fig.transFigure.inverted().transform((xd, yd))

pp = to_fig(ax3, fig2, r_pure)
mp = to_fig(ax3, fig2, r_mixed)
cp = to_fig(ax3, fig2, [0,0,0])
sp = to_fig(ax3, fig2, spiral_pt)

def annotate(fig, target, dx, dy, text, color):
    tx, ty = target
    lx, ly = tx+dx, ty+dy
    fig.text(lx, ly, text, color=color, fontsize=10, ha='center', va='center',
             bbox=dict(boxstyle='round,pad=0.35', fc='#1e1e2e', ec=color, lw=1.2))
    bx = lx - np.sign(dx)*0.055
    by = ly - np.sign(dy)*0.040
    fig.add_artist(mlines.Line2D([bx,tx],[by,ty],
        transform=fig.transFigure, color=color, lw=1.0, ls='--', alpha=0.75))

D = 0.14
annotate(fig2, pp,  +D,      +D,   'Pure state\n$|\\mathbf{r}|=1$',  BLUE)
annotate(fig2, mp,  +D,      -D,   'Mixed state\n$|\\mathbf{r}|<1$', YELLOW)
annotate(fig2, cp,  -D,   -0.04,   'Max. mixed\n$\\rho=I/2$',        RED)
annotate(fig2, sp,  -D*2.2,  +D,   'Decoherence\n(dephasing)',          GREEN)

# 2nd draw → read canvas buffer directly (no savefig!)
fig2.canvas.draw()
buf = fig2.canvas.buffer_rgba()
w, h = fig2.canvas.get_width_height()
img = Image.frombuffer('RGBA', (w, h), buf, 'raw', 'RGBA', 0, 1).convert('RGB')

# Splice out blank gap between title and sphere
arr = np.array(img)
row_density = (np.abs(arr.astype(int) - BG_RGB).max(axis=2) > 8).sum(axis=1)

title_end    = max((r for r in range(min(80, len(row_density))) if row_density[r] > 5), default=30)
sphere_start = next((r for r in range(title_end+1, len(row_density)) if row_density[r] > 5), title_end+1)

PAD = 10
title_part  = img.crop((0, 0,                  img.width, title_end + PAD))
sphere_part = img.crop((0, sphere_start - PAD,  img.width, img.height))

# Also trim bottom blank space
arr2 = np.array(sphere_part)
rd2  = (np.abs(arr2.astype(int) - BG_RGB).max(axis=2) > 8).sum(axis=1)
bot  = max((r for r in range(len(rd2)) if rd2[r] > 5), default=len(rd2)-1)
sphere_part = sphere_part.crop((0, 0, sphere_part.width, bot + PAD))

result = Image.new('RGB', (img.width, title_part.height + sphere_part.height), (30,30,46))
result.paste(title_part,  (0, 0))
result.paste(sphere_part, (0, title_part.height))

path = f'{OUTDIR}/plot5b_bloch_sphere.png'
result.save(path, dpi=(150,150))
print(f'  saved -> {path}  ({result.size[0]}x{result.size[1]}px)')
plt.show()
plt.close()
```

## Plot 6 · Ch 9 — Bell Inequality & CHSH Violation

**Right**: Entanglement entropy $S_{\rm ent}$ vs Schmidt parameter $\alpha$ for $|\psi\rangle=\cos\alpha|00\rangle+\sin\alpha|11\rangle$.

```
# ── Check output dir exists ─────────────────────────────
import os; os.makedirs(OUTDIR, exist_ok=True)
print(f"Saving to: {os.path.abspath(OUTDIR)}")
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import Circle, FancyArrowPatch
from mpl_toolkits.mplot3d import Axes3D
import os, zipfile

BG='#1e1e2e'; FG='#cdd6f4'; EDGE='#45475a'
BLUE='#7ec8e3'; YELLOW='#f9c74f'; GREEN='#90ee90'
RED='#e07070'; PURPLE='#c792ea'; ORANGE='#ffb347'

plt.rcParams.update({
    'figure.facecolor':BG,'axes.facecolor':BG,
    'text.color':FG,'axes.labelcolor':FG,
    'xtick.color':FG,'ytick.color':FG,
    'axes.edgecolor':EDGE,'grid.color':EDGE,'grid.alpha':0.28,
    'font.family':'DejaVu Sans','font.size':12,
    'axes.titlesize':13,'axes.titleweight':'bold',
    'axes.titlecolor':BLUE,'legend.framealpha':0.3,
    'legend.edgecolor':EDGE,
})

OUTDIR = 'aqm_figures'
os.makedirs(OUTDIR, exist_ok=True)

def save(name, fig=None):
    f = fig or plt.gcf()
    path = f'{OUTDIR}/{name}'
    f.savefig(path, dpi=180, bbox_inches='tight', facecolor=BG)
    print(f'  saved -> {path}')
    plt.show()


fig, axes = plt.subplots(1, 3, figsize=(13, 5))
fig.patch.set_facecolor(BG)

# ── Panel 1: Correlation E(θ) ─────────────────────────────────────
ax = axes[0]
theta = np.linspace(0, np.pi, 500)
theta_deg = np.degrees(theta)

E_QM  = -np.cos(theta)
E_LHV = 1 - 2*theta/np.pi   # best deterministic LHV (linear)

ax.plot(theta_deg, E_QM,  color=BLUE, lw=2.5, label=r'QM: $E=-\cos\theta$')
ax.plot(theta_deg, E_LHV, color=RED,  lw=2.0, ls='--', label='Best LHV (linear)')
ax.fill_between(theta_deg, E_LHV, E_QM,
                where=E_QM < E_LHV, color=BLUE, alpha=0.13,
                label='QM stronger (↓ = more negative)')
ax.fill_between(theta_deg, E_LHV, E_QM,
                where=E_QM > E_LHV, color=RED, alpha=0.08)

# Optimal CHSH angles
for ang, lab in [(45,r'$45°$'),(90,r'$90°$'),(135,r'$135°$')]:
    ax.axvline(ang, color=YELLOW, lw=0.8, ls=':', alpha=0.7)
    ax.text(ang+2, -0.95, lab, color=YELLOW, fontsize=8)

ax.axhline(0, color=EDGE, lw=0.5)
ax.set_xlabel(r'Angle $\theta$ (degrees)', fontsize=11)
ax.set_ylabel(r'Correlation $E(\theta)$', fontsize=11)
ax.set_title('QM vs LHV Correlations\n(Singlet state)', color=BLUE)
ax.legend(fontsize=8.5, labelcolor=FG)
ax.set_xlim(0, 180); ax.set_ylim(-1.12, 1.12)
ax.grid(True)

# ── Panel 2: CHSH S vs angle ──────────────────────────────────────
ax2 = axes[1]
phi = np.linspace(0, np.pi/2, 400)
# Optimal CHSH with a=0, a'=π/2, b=φ, b'=φ−π/2
S_QM = np.abs(2*np.sqrt(2)*np.cos(2*phi - np.pi/4))

ax2.plot(np.degrees(phi), S_QM, color=BLUE, lw=2.5, label=r'$S_{\rm QM}(\phi)$')
ax2.axhline(2,             color=RED,   lw=2.0, ls='--', label=r'LHV bound $S\leq 2$')
ax2.axhline(2*np.sqrt(2),  color=GREEN, lw=1.5, ls=':',
            label=r'Tsirelson: $2\sqrt{2}\approx2.83$')
ax2.fill_between(np.degrees(phi), 2, S_QM, where=S_QM>2,
                 color=BLUE, alpha=0.15, label='Quantum violation')

# Optimal angle
ax2.axvline(22.5, color=YELLOW, lw=1.0, ls='--', alpha=0.8)
ax2.text(24, 2.1, r'Optimal $22.5°$', color=YELLOW, fontsize=8.5)
ax2.text(24, 2*np.sqrt(2)+0.05, r'$2\sqrt{2}$', color=GREEN, fontsize=9)

ax2.set_xlabel(r'Detector angle $\phi$ (degrees)', fontsize=11)
ax2.set_ylabel(r'CHSH value $S$', fontsize=11)
ax2.set_title('CHSH Violation\n(Nobel Prize 2022)', color=BLUE)
ax2.legend(fontsize=8.5, labelcolor=FG)
ax2.set_xlim(0, 90); ax2.set_ylim(0, 3.2)
ax2.grid(True)

# ── Panel 3: Entanglement entropy ────────────────────────────────
ax3 = axes[2]
alpha = np.linspace(1e-6, np.pi/2-1e-6, 500)
lam1 = np.cos(alpha)**2
lam2 = np.sin(alpha)**2
S_ent = -(lam1*np.log(lam1+1e-15) + lam2*np.log(lam2+1e-15))

ax3.plot(np.degrees(alpha), S_ent, color=BLUE, lw=2.5)
ax3.fill_between(np.degrees(alpha), 0, S_ent, color=BLUE, alpha=0.12)

# Key points
ax3.plot(0,  0, 'o', color=GREEN, ms=9, zorder=5,
         label=r'$\alpha=0$: product $|00\rangle$, $S=0$')
ax3.plot(45, np.log(2), 'o', color=YELLOW, ms=9, zorder=5,
         label=r'$\alpha=\pi/4$: Bell state, $S=\ln2$')
ax3.axhline(np.log(2), color=YELLOW, lw=0.8, ls='--', alpha=0.6)
ax3.text(47, np.log(2)+0.02, r'$\ln 2$ (max)', color=YELLOW, fontsize=9)

ax3.set_xlabel(r'Parameter $\alpha$ (degrees)', fontsize=11)
ax3.set_ylabel(r'Entanglement entropy $S_{\rm ent}$', fontsize=11)
ax3.set_title(
    r'Entanglement Entropy of $\cos\alpha|00\rangle+\sin\alpha|11\rangle$',
    color=BLUE, fontsize=10)
ax3.legend(fontsize=8.5, labelcolor=FG)
ax3.set_xlim(0, 90); ax3.set_ylim(-0.02, 0.80)
ax3.grid(True)

plt.tight_layout(pad=1.2)
save('plot6_bell_entanglement.png', fig)
plt.close()
```

## Plot 7 · Ch 10 — Scattering Theory

**Left**: Partial wave phase shift $\delta_0$ and cross-section $\sigma_0 \propto \sin^2\delta_0$ vs energy — Breit–Wigner resonance.  
**Centre**: Born approximation $d\sigma/d\Omega$ for Yukawa potential vs angle $\theta$ for several screening lengths.  
**Right**: Bragg diffraction peaks $|S(q)|^2$ and Debye–Waller suppression at finite temperature.

```
# ── Check output dir exists ─────────────────────────────
import os; os.makedirs(OUTDIR, exist_ok=True)
print(f"Saving to: {os.path.abspath(OUTDIR)}")
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import Circle, FancyArrowPatch
from mpl_toolkits.mplot3d import Axes3D
import os, zipfile

BG='#1e1e2e'; FG='#cdd6f4'; EDGE='#45475a'
BLUE='#7ec8e3'; YELLOW='#f9c74f'; GREEN='#90ee90'
RED='#e07070'; PURPLE='#c792ea'; ORANGE='#ffb347'

plt.rcParams.update({
    'figure.facecolor':BG,'axes.facecolor':BG,
    'text.color':FG,'axes.labelcolor':FG,
    'xtick.color':FG,'ytick.color':FG,
    'axes.edgecolor':EDGE,'grid.color':EDGE,'grid.alpha':0.28,
    'font.family':'DejaVu Sans','font.size':12,
    'axes.titlesize':13,'axes.titleweight':'bold',
    'axes.titlecolor':BLUE,'legend.framealpha':0.3,
    'legend.edgecolor':EDGE,
})

OUTDIR = 'aqm_figures'
os.makedirs(OUTDIR, exist_ok=True)

def save(name, fig=None):
    f = fig or plt.gcf()
    path = f'{OUTDIR}/{name}'
    f.savefig(path, dpi=180, bbox_inches='tight', facecolor=BG)
    print(f'  saved -> {path}')
    plt.show()


fig, axes = plt.subplots(1, 3, figsize=(13, 5))
fig.patch.set_facecolor(BG)

# ── Panel 1: Breit–Wigner resonance ──────────────────────────────
ax = axes[0]
E_arr = np.linspace(0.01, 3.0, 800)
E_R, Gamma_BW = 1.0, 0.35   # resonance energy and width

# Phase shift: arctan[(Γ/2)/(E_R−E)], with π step at resonance
delta0 = np.arctan((Gamma_BW/2) / (E_R - E_arr + 1e-10))
delta0[E_arr > E_R] += np.pi

sigma0 = np.sin(delta0)**2   # ∝ σ₀ (normalised)

ax_twin = ax.twinx()
ax_twin.set_facecolor(BG)
ax_twin.tick_params(colors=FG)
ax_twin.spines['right'].set_color(GREEN)

ax.plot(E_arr, np.degrees(delta0), color=BLUE, lw=2.2, label=r'$\delta_0(E)$')
ax_twin.plot(E_arr, sigma0, color=GREEN, lw=2.0, ls='--',
             label=r'$\sin^2\delta_0$ (norm. $\sigma_0$)')

ax.axvline(E_R, color=RED, lw=1.2, ls='--', alpha=0.8)
ax.axhline(90,  color=YELLOW, lw=0.8, ls=':', alpha=0.7)
ax.text(E_R+0.05, 15, r'$E_R$', color=RED, fontsize=10)
ax.text(0.08, 93, r'$\delta_0=90°$', color=YELLOW, fontsize=8.5)

# Γ annotation
half = np.where(sigma0 >= 0.5)[0]
if len(half) > 1:
    ax_twin.hlines(0.5, E_arr[half[0]], E_arr[half[-1]],
                   colors=RED, lw=1.2, ls='--')
    ax_twin.annotate('', xy=(E_arr[half[-1]], 0.52),
                     xytext=(E_arr[half[0]], 0.52),
                     arrowprops=dict(arrowstyle='<->', color=RED,
                                     lw=1.2, mutation_scale=10))
    ax_twin.text((E_arr[half[0]]+E_arr[half[-1]])/2, 0.57,
                 r'$\Gamma$', color=RED, ha='center', fontsize=11)

ax.set_xlabel('Energy $E$', fontsize=11)
ax.set_ylabel(r'Phase shift $\delta_0$ (degrees)', fontsize=11)
ax_twin.set_ylabel(r'$\sin^2\delta_0$', fontsize=10, color=GREEN)
ax.set_title('Breit–Wigner Resonance\n(s-wave scattering)', color=BLUE)
lines1,labs1 = ax.get_legend_handles_labels()
lines2,labs2 = ax_twin.get_legend_handles_labels()
ax.legend(lines1+lines2, labs1+labs2, fontsize=9, labelcolor=FG, loc='center right')
ax.set_xlim(0, 3); ax.set_ylim(0, 195)
ax_twin.set_ylim(-0.05, 1.12)
ax.grid(True)

# ── Panel 2: Born approx Yukawa dσ/dΩ ───────────────────────────
ax2 = axes[1]
theta_arr = np.linspace(1, 180, 500)
theta_r   = np.radians(theta_arr)
k_inc = 3.0   # incident wavevector
q_arr = 2*k_inc*np.sin(theta_r/2)

for mu, col, lab in [(0.3, BLUE, r'$\mu=0.3$ (long range)'),
                      (1.5, YELLOW, r'$\mu=1.5$'),
                      (5.0, GREEN,  r'$\mu=5.0$ (short range)')]:
    dsigma = 1.0 / (q_arr**2 + mu**2)**2
    dsigma /= dsigma[0]   # normalise to forward value
    ax2.semilogy(theta_arr, dsigma, color=col, lw=2.2, label=lab)

# Rutherford (μ→0) reference
dsigma_R = 1.0 / np.sin(theta_r/2)**4
dsigma_R /= dsigma_R[0]
ax2.semilogy(theta_arr, dsigma_R, color=RED, lw=1.5, ls='--',
             alpha=0.8, label=r'Rutherford ($\mu\to 0$)')

ax2.set_xlabel(r'Scattering angle $\theta$ (degrees)', fontsize=11)
ax2.set_ylabel(r'$d\sigma/d\Omega$ (normalised)', fontsize=11)
ax2.set_title('Born Approx: Yukawa Potential\n$d\\sigma/d\\Omega$', color=BLUE, fontsize=10)
ax2.legend(fontsize=8.5, labelcolor=FG)
ax2.set_xlim(0, 180)
ax2.grid(True)

# ── Panel 3: Bragg peaks + Debye–Waller ─────────────────────────
ax3 = axes[2]
N  = 14    # number of lattice sites
q  = np.linspace(-9*np.pi, 9*np.pi, 5000)  # q in units of 1/a
qa2 = q / 2.0

with np.errstate(divide='ignore', invalid='ignore'):
    S2 = np.where(np.abs(np.sin(qa2)) < 1e-7,
                  float(N)**2,
                  np.sin(N*qa2)**2 / np.sin(qa2)**2)
S2_norm = S2 / N**2

# Debye–Waller factors
u2_vals  = [0.0, 0.02, 0.08]   # ⟨u²⟩ in units of a²
DW_cols  = [BLUE, YELLOW, RED]
DW_labs  = ['$T=0$', '$T=T_1$ (warm)', '$T=T_2$ (hot)']

for u2, col, lab in zip(u2_vals, DW_cols, DW_labs):
    DW = np.exp(-q**2 * u2)
    ax3.plot(q/np.pi, S2_norm * DW, color=col, lw=1.8, alpha=0.9, label=lab)

# BZ boundary markers
for n in range(-4, 5):
    if n != 0:
        ax3.axvline(2*n, color=EDGE, lw=0.6, ls=':', alpha=0.5)

ax3.text(2.2, 0.78, r'Debye–Waller:\n$I\propto e^{-q^2\langle u^2\rangle}$',
         color=FG, fontsize=8,
         bbox=dict(boxstyle='round,pad=0.3', fc=BG, ec=EDGE, lw=0.8))

ax3.set_xlabel(r'$qa/\pi$', fontsize=11)
ax3.set_ylabel(r'$|S(q)|^2/N^2$ × DW factor', fontsize=10)
ax3.set_title(f'Bragg Peaks + Debye–Waller\n($N={N}$ sites)', color=BLUE)
ax3.legend(fontsize=9, labelcolor=FG)
ax3.set_xlim(-9, 9); ax3.set_ylim(-0.02, 1.05)
ax3.set_xticks([-8,-6,-4,-2,0,2,4,6,8])
ax3.grid(True)

plt.tight_layout(pad=1.2)
save('plot7_scattering.png', fig)
plt.close()
```
